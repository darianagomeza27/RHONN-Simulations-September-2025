# Double Pendulum Identification with RHONN

## Overview
Neural identification using three training algorithms:
- **EKF-RHONN**: Extended Kalman Filter
- **UKF-RHONN**: Unscented Kalman Filter  
- **PF-RHONN**: Particle Filter (neuron-specific parameters)

## Features
- RK4 discretization
- Non-Gaussian noise models
- NaN prevention strategies
- Comprehensive error metrics

## Execution
Run cells in order: Definitions → Simulation → Visualizations

```python
# Neural Identifier Training - Double Pendulum
# Methods: EKF, UKF, Particle Filter
# Con características específicas por neurona

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.optimize import minimize, differential_evolution
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# 1) True nonlinear system (Double Pendulum)
# ============================================================
def plant_dynamics(state, u):
    """
    Dynamics of a double pendulum with damping and external torques.
    state = [θ1, θ2, ω1, ω2] 
        θ1, θ2: angles (rad)
        ω1, ω2: angular velocities (rad/s)
    u = [τ1, τ2]: external torques on joints (N⋅m)
    """
    # Physical Parameters
    m1 = 1.0     # Mass of first pendulum (kg)
    m2 = 1.0     # Mass of second pendulum (kg)
    l1 = 1.0     # Length of first pendulum (m)
    l2 = 1.0     # Length of second pendulum (m)
    g = 9.81     # Gravity (m/s²)
    b1 = 0.1     # Damping coefficient for joint 1 (N⋅m⋅s)
    b2 = 0.1     # Damping coefficient for joint 2 (N⋅m⋅s)
    
    θ1, θ2, ω1, ω2 = state
    τ1, τ2 = u
    
    # Precompute trigonometric terms
    c12 = np.cos(θ1 - θ2)  # cos(θ1 - θ2)
    s12 = np.sin(θ1 - θ2)  # sin(θ1 - θ2)
    s1 = np.sin(θ1)         # sin(θ1)
    s2 = np.sin(θ2)         # sin(θ2)
    
    # Mass matrix elements
    M11 = (m1 + m2) * l1**2
    M12 = m2 * l1 * l2 * c12
    M21 = M12
    M22 = m2 * l2**2
    
    # Coriolis and centrifugal terms
    h = m2 * l1 * l2 * ω2**2 * s12
    C1 = h - (m1 + m2) * g * l1 * s1 - b1 * ω1 + τ1
    
    C2 = -m2 * l1 * l2 * ω1**2 * s12 - m2 * g * l2 * s2 - b2 * ω2 + τ2
    
    # Solve for accelerations using the mass matrix
    # [M11  M12] [α1]   [C1]
    # [M21  M22] [α2] = [C2]
    det_M = M11 * M22 - M12 * M21
    
    if abs(det_M) < 1e-10:
        det_M = 1e-10  # Prevent division by zero
    
    α1 = (M22 * C1 - M12 * C2) / det_M
    α2 = (-M21 * C1 + M11 * C2) / det_M
    
    # State derivatives
    θ1_dot = ω1
    θ2_dot = ω2
    ω1_dot = α1
    ω2_dot = α2
    
    return np.array([θ1_dot, θ2_dot, ω1_dot, ω2_dot])

def plant(x_k, u_k, dt=0.01):
    """
    RK4 integration step for better accuracy.
    """
    # RK4 para mayor precisión
    k1 = plant_dynamics(x_k, u_k)
    k2 = plant_dynamics(x_k + 0.5*dt*k1, u_k)
    k3 = plant_dynamics(x_k + 0.5*dt*k2, u_k)
    k4 = plant_dynamics(x_k + dt*k3, u_k)
    
    x_kp1 = x_k + (dt/6.0) * (k1 + 2*k2 + 2*k3 + k4)
    
    return x_kp1

# ============================================================
# 2) RHONN structure - CARACTERÍSTICAS POR NEURONA
# ============================================================
def sigmoidal(z, beta=1.0):
    """Sigmoid S(z)."""
    z = np.clip(z, -50, 50) 
    return 1.0 / (1.0 + np.exp(-beta * z))

def construct_z_vector(x_est, u_input, neuron_index):
    """
    Features for Double Pendulum - ESPECÍFICAS PARA CADA NEURONA.
    
    x_est = [θ1, θ2, ω1, ω2]
    u_input = [τ1, τ2]
    neuron_index: índice de la neurona (0=θ1, 1=θ2, 2=ω1, 3=ω2)
    """
    θ1, θ2, ω1, ω2 = x_est
    τ1, τ2 = u_input
    
    # Términos básicos sigmoidales
    s_θ1 = sigmoidal(θ1)
    s_θ2 = sigmoidal(θ2)
    s_ω1 = sigmoidal(ω1)
    s_ω2 = sigmoidal(ω2)
    
    # Términos trigonométricos (importantes para la dinámica del péndulo)
    # cos_θ1 = np.cos(θ1)
    # sin_θ1 = np.sin(θ1)
    # cos_θ2 = np.cos(θ2)
    # sin_θ2 = np.sin(θ2)
    # cos_θ12 = np.cos(θ1 - θ2)
    # sin_θ12 = np.sin(θ1 - θ2)
    
    # Comandos escalados
    s_τ1 = sigmoidal(τ1)
    s_τ2 = sigmoidal(τ2)
    
    # ========== CARACTERÍSTICAS ESPECÍFICAS POR NEURONA ==========
    
    if neuron_index == 0:  # Neurona para θ1 (ángulo del primer péndulo)
        # dθ1/dt = ω1
        return np.array([
            # s_ω1,                      # Velocidad angular (término principal)
            s_ω1**3,                   # Término cuadrático
            s_θ1**2,                      # Posición angular
            s_θ1 * s_ω1,              # Acoplamiento posición-velocidad
            # s_ω2,                      # Acoplamiento con segundo péndulo
            # cos_θ12,                   # Término de acoplamiento geométrico
        ])
    
    elif neuron_index == 1:  # Neurona para θ2 (ángulo del segundo péndulo)
        # dθ2/dt = ω2
        return np.array([
            s_ω2,                      # Velocidad angular (término principal)
            s_ω2**2,                   # Término cuadrático
            # s_θ2,                      # Posición angular
            s_θ2 * s_ω2,              # Acoplamiento posición-velocidad
            s_ω1,                      # Acoplamiento con primer péndulo
            # cos_θ12,                   # Término de acoplamiento geométrico
        ])
    
    elif neuron_index == 2:  # Neurona para ω1 (velocidad angular del primer péndulo)
        # dω1/dt = función compleja de θ1, θ2, ω1, ω2, τ1
        return np.array([
            s_ω1,                      # Estado actual
            s_ω1**2,                   # Término cuadrático (fricción)
            # sin_θ1,                    # Término gravitacional
            # sin_θ12,                   # Acoplamiento con segundo péndulo
            # s_ω2 * sin_θ12,           # Término de Coriolis
            s_τ1,                      # Torque externo
            # cos_θ12,                   # Acoplamiento geométrico
            s_ω1 * s_ω2,              # Interacción velocidades
        ])
    
    elif neuron_index == 3:  # Neurona para ω2 (velocidad angular del segundo péndulo)
        # dω2/dt = función compleja de θ1, θ2, ω1, ω2, τ2
        return np.array([
            s_ω2,                      # Estado actual
            s_ω2**2,                   # Término cuadrático (fricción)
            # sin_θ2,                    # Término gravitacional
            # sin_θ12,                   # Acoplamiento con primer péndulo
            # s_ω1 * sin_θ12,           # Término de Coriolis
            s_τ2,                      # Torque externo
            # cos_θ12,                   # Acoplamiento geométrico
            s_ω1 * s_ω2,              # Interacción velocidades
        ])
    
    else:
        raise ValueError(f"Índice de neurona inválido: {neuron_index}")

# Función auxiliar para obtener el tamaño de características de cada neurona
def get_z_size(neuron_index):
    """Retorna el número de características para una neurona dada."""
    if neuron_index == 0:  # θ1
        return 3
    elif neuron_index == 1:  # θ2
        return 4
    elif neuron_index == 2:  # ω1
        return 4
    elif neuron_index == 3:  # ω2
        return 4
    else:
        raise ValueError(f"Índice de neurona inválido: {neuron_index}")

# ============================================================
# 3) Trainers (EKF, UKF, PF) - ADAPTADOS PARA MÚLTIPLES TAMAÑOS
# ============================================================

class Generic_RHONN_Trainer:
    """ Base class to handle the loop logic easily """
    def get_prediction(self, weights, x_k, u_k, neuron_idx):
        z = construct_z_vector(x_k, u_k, neuron_idx)
        return np.dot(weights, z)

class EKF_Trainer(Generic_RHONN_Trainer):
    '''
    eta = 0.9780
   P0 = 3.9958
   Q = 1.24e-03
   R = 2.46e-04
    '''
    def __init__(self, n_neurons, eta=1.0, P0=1.0, Q=1e-3, R=1e-5):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(get_z_size(i))*0.1 for i in range(n_neurons)]
        self.P = [np.eye(get_z_size(i))*P0 for i in range(n_neurons)]
        self.Q_matrices = [np.eye(get_z_size(i))*Q for i in range(n_neurons)]
        self.R = R
        self.eta = eta

    def update(self, x_kp1, x_k, u_k):
        for i in range(self.n_neurons):
            # Construir vector de características H_i (ecuación 8)
            z = construct_z_vector(x_k, u_k, i)
            H_i = z.reshape(-1, 1)
            
            # Error de identificación e_i(k) (ecuación 7)
            # e_i(k) = x_i(k) - X̂_i(k)
            x_hat_i = np.dot(self.weights[i], z)
            e_i = x_kp1[i] - x_hat_i
            
            # Ganancia de Kalman K_i(k) (ecuación 6)
            # K_i(k) = P_i(k) H_i(k) [R_i(k) + H_i(k) P_i(k) H_i(k)]^{-1}
            S = self.R + (H_i.T @ self.P[i] @ H_i)[0, 0]
            K_i = (self.P[i] @ H_i).flatten() / S
            
            # Actualización de pesos ω_i(k+1) (ecuación superior)
            # ω_i(k+1) = ω_i(k) + η_i K_i(k) e_i(k)
            self.weights[i] = self.weights[i] + self.eta * K_i * e_i
            
            # Actualización de covarianza P_i(k+1) (ecuación 6, tercera línea)
            # P_i(k+1) = P_i(k) - K_i(k) H_i(k) P_i(k) + Q_i(k)
            self.P[i] = self.P[i] - np.outer(K_i, H_i.flatten()) @ self.P[i] + self.Q_matrices[i]

class UKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, eta=1.0, alpha=1e-2, Q_std=1e-3, R_std=1e-5):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(get_z_size(i))*0.1 for i in range(n_neurons)]
        self.P = [np.eye(get_z_size(i)) for i in range(n_neurons)]
        self.Q = Q_std
        self.Q_matrices = [np.eye(get_z_size(i))*Q_std for i in range(n_neurons)]
        self.R = R_std
        self.eta = eta
        self.alpha = alpha

    def update(self, x_kp1, x_k, u_k):
        for i in range(self.n_neurons):
            z = construct_z_vector(x_k, u_k, i)
            n = get_z_size(i)
            
            # Sigma params
            lambda_ = self.alpha**2 * n - n
            Wm = np.full(2*n+1, 1/(2*(n+lambda_)))
            Wc = np.copy(Wm)
            Wm[0] = lambda_/(n+lambda_)
            Wc[0] = Wm[0] + (3 - self.alpha**2)
            
            # Generate Sigmas
            try:
                L = np.linalg.cholesky((n + lambda_) * self.P[i])
            except:
                L = np.eye(n) * 0.1
                
            sigmas = np.zeros((2*n+1, n))
            sigmas[0] = self.weights[i]
            for k in range(n):
                sigmas[k+1] = self.weights[i] + L[:,k]
                sigmas[n+k+1] = self.weights[i] - L[:,k]
            
            # Transform
            Y_sigmas = np.dot(sigmas, z)
            y_mean = np.sum(Wm * Y_sigmas)
            
            # Covariances
            Py = np.sum(Wc * (Y_sigmas - y_mean)**2) + self.R
            Pxy = np.zeros(n)
            for k in range(2*n+1):
                Pxy += Wc[k] * (sigmas[k] - self.weights[i]) * (Y_sigmas[k] - y_mean)
                
            # Update
            K = Pxy / Py
            err = x_kp1[i] - y_mean
            self.weights[i] += self.eta * K * err
            self.P[i] -= np.outer(K, K) * Py
            
            # Regularize P
            self.P[i] += np.eye(n)*1e-6

class PF_Trainer(Generic_RHONN_Trainer):
    """
    Robust Particle Filter Trainer with NaN prevention strategies.
    Now supports neuron-specific Q and R values.
    """
    def __init__(self, n_neurons, n_particles=1000, Q_std=None, R_std=None):
        self.n_neurons = n_neurons
        self.n_particles = n_particles
        
        # Initialize particles with controlled variance
        self.particles = [np.random.randn(n_particles, get_z_size(i)) * 0.1 
                         for i in range(n_neurons)]
        
        # Initialize weights uniformly
        self.weights_pf = [np.ones(n_particles) / n_particles for _ in range(n_neurons)]
        
        # Tuning parameters - neuron-specific or default
        if Q_std is None:
            # Default: same for all neurons
            self.Q_std = [0.05] * n_neurons
        elif isinstance(Q_std, (list, np.ndarray)):
            # Use provided neuron-specific values
            assert len(Q_std) == n_neurons, f"Q_std must have length {n_neurons}"
            self.Q_std = list(Q_std)
        else:
            # Single value for all neurons
            self.Q_std = [Q_std] * n_neurons
        
        if R_std is None:
            # Default: same for all neurons
            self.R_std = [0.05] * n_neurons
        elif isinstance(R_std, (list, np.ndarray)):
            # Use provided neuron-specific values
            assert len(R_std) == n_neurons, f"R_std must have length {n_neurons}"
            self.R_std = list(R_std)
        else:
            # Single value for all neurons
            self.R_std = [R_std] * n_neurons
        
        self.regularization_std = 0.001  # Jitter after resampling
        
        # NaN prevention parameters
        self.min_weight = 1e-300    # Minimum weight to prevent underflow
        self.max_log_likelihood = 10000.0  # Clip extreme likelihoods
        self.resample_threshold = 0.5   # Resample when Neff < threshold * N
        
    def _normalize_weights(self, weights):
        """
        Safely normalize weights with NaN and underflow protection.
        """
        # Check for NaN or inf
        if np.any(~np.isfinite(weights)):
            print("⚠️  Warning: Non-finite weights detected, resetting to uniform")
            return np.ones_like(weights) / len(weights)
        
        # Ensure positive weights
        weights = np.maximum(weights, self.min_weight)
        
        # Normalize
        weight_sum = np.sum(weights)
        if weight_sum < self.min_weight or not np.isfinite(weight_sum):
            print("⚠️  Warning: Invalid weight sum, resetting to uniform")
            return np.ones_like(weights) / len(weights)
        
        return weights / weight_sum
    
    def _compute_log_likelihood(self, errors, neuron_idx):
        """
        Compute log-likelihood with numerical stability.
        Uses neuron-specific R_std.
        """
        # Clip extreme errors
        errors_clipped = np.clip(errors, -1000, 1000)
        
        # Gaussian log-likelihood: -0.5 * (err/sigma)^2
        # sigma = self.R_std[neuron_idx]
        # normalized_errors_sq = (errors_clipped / sigma) ** 2
        # log_likelihood = -0.5 * normalized_errors_sq
        
        # Student-t log-likelihood with df=3 (commented out)
        df = 2.0  # degrees of freedom (heavy tails)
        sigma = self.R_std[neuron_idx]
        normalized_errors_sq = (errors_clipped / sigma) ** 2
        log_likelihood = -((df + 1) / 2.0) * np.log(1.0 + normalized_errors_sq / df)
        
        return log_likelihood
    
    def _resample_particles(self, neuron_idx):
        """
        Systematic resampling with regularization.
        """
        weights = self.weights_pf[neuron_idx]
        particles = self.particles[neuron_idx]
        n_weights = get_z_size(neuron_idx)
        
        # Normalize weights
        weights = self._normalize_weights(weights)
        
        # Systematic resampling
        positions = (np.arange(self.n_particles) + np.random.random()) / self.n_particles
        cumulative_sum = np.cumsum(weights)
        
        indices = np.searchsorted(cumulative_sum, positions)
        indices = np.clip(indices, 0, self.n_particles - 1)
        
        # Resample particles
        self.particles[neuron_idx] = particles[indices].copy()
        
        # Add regularization jitter (preserve particle diversity)
        jitter = np.random.randn(self.n_particles, n_weights) * self.regularization_std
        self.particles[neuron_idx] += jitter
        
        # Reset weights to uniform
        self.weights_pf[neuron_idx] = np.ones(self.n_particles) / self.n_particles
    
    def update(self, x_kp1, x_k, u_k):
        """
        Update step with NaN prevention and neuron-specific Q/R.
        """
        for i in range(self.n_neurons):
            try:
                # Get feature vector
                z = construct_z_vector(x_k, u_k, i)
                n_weights = get_z_size(i)
                
                # Check for NaN in input
                if not np.all(np.isfinite(z)):
                    print(f"⚠️  Warning: Non-finite feature vector for neuron {i}, skipping update")
                    continue
                
                # 1. PREDICTION: Add process noise (drift) - neuron-specific Q_std
                drift = np.random.randn(self.n_particles, n_weights) * self.Q_std[i]
                self.particles[i] += drift
                
                # Clip particles to prevent extreme values
                self.particles[i] = np.clip(self.particles[i], -1000, 1000)
                
                # 2. UPDATE: Compute likelihoods
                preds = self.particles[i] @ z
                
                # Check for NaN in predictions
                if not np.all(np.isfinite(preds)):
                    print(f"⚠️  Warning: Non-finite predictions for neuron {i}, resetting particles")
                    self.particles[i] = np.random.randn(self.n_particles, n_weights) * 0.1
                    continue
                
                # Compute errors
                errors = x_kp1[i] - preds
                
                # Compute log-likelihood (numerically stable) - neuron-specific R_std
                log_likelihood = self._compute_log_likelihood(errors, i)
                
                # Shift for numerical stability before exp
                log_likelihood -= np.max(log_likelihood)
                
                # Update weights
                self.weights_pf[i] *= np.exp(log_likelihood)
                
                # Normalize weights
                self.weights_pf[i] = self._normalize_weights(self.weights_pf[i])
                
                # 3. RESAMPLING: Check effective sample size
                weight_sq_sum = np.sum(self.weights_pf[i] ** 2)
                if weight_sq_sum > 0:
                    eff_N = 1.0 / weight_sq_sum
                else:
                    eff_N = 0
                
                # Resample if effective sample size is too low
                if eff_N < self.resample_threshold * self.n_particles:
                    self._resample_particles(i)
                    
            except Exception as e:
                print(f"⚠️  Error in PF update for neuron {i}: {e}")
                # Reset to safe state
                self.particles[i] = np.random.randn(self.n_particles, get_z_size(i)) * 0.1
                self.weights_pf[i] = np.ones(self.n_particles) / self.n_particles
    
    def get_estimates(self):
        """
        Get weighted average of particles (with NaN protection).
        """
        estimates = []
        for i in range(self.n_neurons):
            # Normalize weights
            weights = self._normalize_weights(self.weights_pf[i])
            
            # Compute weighted average
            estimate = np.average(self.particles[i], axis=0, weights=weights)
            
            # Check for NaN
            if not np.all(np.isfinite(estimate)):
                print(f"⚠️  Warning: Non-finite estimate for neuron {i}, using median")
                estimate = np.median(self.particles[i], axis=0)
            
            estimates.append(estimate)
        
        return estimates

# ============================================================
# 4) Error Metrics Functions
# ============================================================
def calculate_error_metrics(y_true, y_pred, metric_name="State"):
    """
    Calculate comprehensive error metrics.
    
    Args:
        y_true: True values (n_samples, n_states)
        y_pred: Predicted values (n_samples, n_states)
        metric_name: Name for reporting
    
    Returns:
        Dictionary with error metrics
    """
    errors = y_true - y_pred
    
    # Mean Absolute Error (MAE)
    mae = np.mean(np.abs(errors), axis=0)
    mae_total = np.mean(mae)
    
    # Root Mean Square Error (RMSE)
    rmse = np.sqrt(np.mean(errors**2, axis=0))
    rmse_total = np.sqrt(np.mean(rmse**2))
    
    # Normalized RMSE (NRMSE) - normalized by range
    ranges = np.max(y_true, axis=0) - np.min(y_true, axis=0)
    ranges[ranges < 1e-10] = 1.0  # Avoid division by zero
    nrmse = rmse / ranges
    nrmse_total = np.mean(nrmse)
    
    # Maximum Absolute Error
    max_error = np.max(np.abs(errors), axis=0)
    
    # Mean Squared Error (MSE)
    mse = np.mean(errors**2, axis=0)
    mse_total = np.mean(mse)
    
    # R² Score (coefficient of determination)
    ss_res = np.sum(errors**2, axis=0)
    ss_tot = np.sum((y_true - np.mean(y_true, axis=0))**2, axis=0)
    r2 = 1 - (ss_res / (ss_tot + 1e-10))
    r2_total = np.mean(r2)
    
    return {
        'MAE': mae,
        'MAE_total': mae_total,
        'RMSE': rmse,
        'RMSE_total': rmse_total,
        'NRMSE': nrmse,
        'NRMSE_total': nrmse_total,
        'MSE': mse,
        'MSE_total': mse_total,
        'Max_Error': max_error,
        'R2': r2,
        'R2_total': r2_total
    }
```

# Double Pendulum Identification with RHONN

## Overview
Neural identification using three training algorithms:
- **EKF-RHONN**: Extended Kalman Filter
- **UKF-RHONN**: Unscented Kalman Filter  
- **PF-RHONN**: Particle Filter (neuron-specific parameters)

## Features
- RK4 discretization
- Non-Gaussian noise models
- NaN prevention strategies
- Comprehensive error metrics

## Execution

```python
# Neural Identifier Training for Double Pendulum System
# Training Methods: EKF, UKF, Particle Filter (with neuron-specific parameters)

import numpy as np
import plotly.graph_objects as go
import warnings
import time

warnings.filterwarnings('ignore')

# ============================================================
# 1) TRUE SYSTEM: Double Pendulum Dynamics
# ============================================================

def plant_dynamics(state, u):
    """
    Dynamics of a double pendulum with damping and external torques.
    state = [θ1, θ2, ω1, ω2] 
    u = [τ1, τ2]: external torques on joints
    """
    # Physical Parameters
    m1, m2 = 1.0, 1.0      # Masses (kg)
    l1, l2 = 1.0, 1.0      # Lengths (m)
    g = 9.81                # Gravity (m/s²)
    b1, b2 = 0.1, 0.1      # Damping coefficients (N⋅m⋅s)
    
    θ1, θ2, ω1, ω2 = state
    τ1, τ2 = u
    
    # Precompute trigonometric terms
    c12 = np.cos(θ1 - θ2)
    s12 = np.sin(θ1 - θ2)
    s1, s2 = np.sin(θ1), np.sin(θ2)
    
    # Mass matrix elements
    M11 = (m1 + m2) * l1**2
    M12 = m2 * l1 * l2 * c12
    M21 = M12
    M22 = m2 * l2**2
    
    # Coriolis and centrifugal terms
    h = m2 * l1 * l2 * ω2**2 * s12
    C1 = h - (m1 + m2) * g * l1 * s1 - b1 * ω1 + τ1
    C2 = -m2 * l1 * l2 * ω1**2 * s12 - m2 * g * l2 * s2 - b2 * ω2 + τ2
    
    # Solve for accelerations
    det_M = M11 * M22 - M12 * M21
    if abs(det_M) < 1e-10:
        det_M = 1e-10  # Prevent division by zero
    
    α1 = (M22 * C1 - M12 * C2) / det_M
    α2 = (-M21 * C1 + M11 * C2) / det_M
    
    # State derivatives
    return np.array([ω1, ω2, α1, α2])

def plant(x_k, u_k, dt=0.01):
    """RK4 integration step."""
    k1 = plant_dynamics(x_k, u_k)
    k2 = plant_dynamics(x_k + 0.5*dt*k1, u_k)
    k3 = plant_dynamics(x_k + 0.5*dt*k2, u_k)
    k4 = plant_dynamics(x_k + dt*k3, u_k)
    
    return x_k + (dt/6.0) * (k1 + 2*k2 + 2*k3 + k4)

# ============================================================
# 2) RHONN STRUCTURE - Neuron-specific features
# ============================================================

def sigmoidal(z, beta=1.0):
    """Sigmoid activation function."""
    z = np.clip(z, -50, 50) 
    return 1.0 / (1.0 + np.exp(-beta * z))

def construct_z_vector(x_est, u_input, neuron_index):
    """
    Features for Double Pendulum - specific for each neuron.
    x_est = [θ1, θ2, ω1, ω2]
    u_input = [τ1, τ2]
    neuron_index: 0=θ1, 1=θ2, 2=ω1, 3=ω2
    """
    θ1, θ2, ω1, ω2 = x_est
    τ1, τ2 = u_input
    
    # Sigmoid terms
    s_θ1 = sigmoidal(θ1)
    s_θ2 = sigmoidal(θ2)
    s_ω1 = sigmoidal(ω1)
    s_ω2 = sigmoidal(ω2)
    s_τ1 = sigmoidal(τ1)
    s_τ2 = sigmoidal(τ2)
    
    # Neuron-specific features
    if neuron_index == 0:  # θ1 (angle of first pendulum)
        return np.array([s_ω1**3, s_θ1**2, s_θ1 * s_ω1])
    
    elif neuron_index == 1:  # θ2 (angle of second pendulum)
        return np.array([s_ω2, s_ω2**2, s_θ2 * s_ω2, s_ω1])
    
    elif neuron_index == 2:  # ω1 (angular velocity of first pendulum)
        return np.array([s_ω1, s_ω1**2, s_τ1, s_ω1 * s_ω2])
    
    elif neuron_index == 3:  # ω2 (angular velocity of second pendulum)
        return np.array([s_ω2, s_ω2**2, s_τ2, s_ω1 * s_ω2])
    
    else:
        raise ValueError(f"Invalid neuron index: {neuron_index}")

def get_z_size(neuron_index):
    """Return number of features for a given neuron."""
    sizes = [3, 4, 4, 4]  # [θ1, θ2, ω1, ω2]
    return sizes[neuron_index]

# ============================================================
# 3) TRAINERS (EKF, UKF, PF) - Adapted for multiple sizes
# ============================================================

class Generic_RHONN_Trainer:
    """Base class for training loop logic."""
    def get_prediction(self, weights, x_k, u_k, neuron_idx):
        z = construct_z_vector(x_k, u_k, neuron_idx)
        return np.dot(weights, z)

class EKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, eta=1.0, P0=1.0, Q=1e-3, R=1e-5):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(get_z_size(i))*0.1 for i in range(n_neurons)]
        self.P = [np.eye(get_z_size(i))*P0 for i in range(n_neurons)]
        self.Q_matrices = [np.eye(get_z_size(i))*Q for i in range(n_neurons)]
        self.R = R
        self.eta = eta

    def update(self, x_kp1, x_k, u_k):
        for i in range(self.n_neurons):
            z = construct_z_vector(x_k, u_k, i)
            H_i = z.reshape(-1, 1)
            
            x_hat_i = np.dot(self.weights[i], z)
            e_i = x_kp1[i] - x_hat_i
            
            # Kalman Gain
            S = self.R + (H_i.T @ self.P[i] @ H_i)[0, 0]
            K_i = (self.P[i] @ H_i).flatten() / S
            
            # Weight update
            self.weights[i] += self.eta * K_i * e_i
            
            # Covariance update
            self.P[i] = self.P[i] - np.outer(K_i, H_i.flatten()) @ self.P[i] + self.Q_matrices[i]

class UKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, eta=1.0, alpha=1e-2, Q_std=1e-3, R_std=1e-5):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(get_z_size(i))*0.1 for i in range(n_neurons)]
        self.P = [np.eye(get_z_size(i)) for i in range(n_neurons)]
        self.Q_matrices = [np.eye(get_z_size(i))*Q_std for i in range(n_neurons)]
        self.R = R_std
        self.eta = eta
        self.alpha = alpha

    def update(self, x_kp1, x_k, u_k):
        for i in range(self.n_neurons):
            z = construct_z_vector(x_k, u_k, i)
            n = get_z_size(i)
            
            # Sigma points parameters
            lambda_ = self.alpha**2 * n - n
            Wm = np.full(2*n+1, 1/(2*(n+lambda_)))
            Wc = np.copy(Wm)
            Wm[0] = lambda_/(n+lambda_)
            Wc[0] = Wm[0] + (3 - self.alpha**2)
            
            # Generate sigma points
            try:
                L = np.linalg.cholesky((n + lambda_) * self.P[i])
            except:
                L = np.eye(n) * 0.1
                
            sigmas = np.zeros((2*n+1, n))
            sigmas[0] = self.weights[i]
            for k in range(n):
                sigmas[k+1] = self.weights[i] + L[:,k]
                sigmas[n+k+1] = self.weights[i] - L[:,k]
            
            # Transform through measurement function
            Y_sigmas = np.dot(sigmas, z)
            y_mean = np.sum(Wm * Y_sigmas)
            
            # Covariances
            Py = np.sum(Wc * (Y_sigmas - y_mean)**2) + self.R
            Pxy = np.zeros(n)
            for k in range(2*n+1):
                Pxy += Wc[k] * (sigmas[k] - self.weights[i]) * (Y_sigmas[k] - y_mean)
                
            # Update
            K = Pxy / Py
            err = x_kp1[i] - y_mean
            self.weights[i] += self.eta * K * err
            self.P[i] -= np.outer(K, K) * Py
            self.P[i] += np.eye(n)*1e-6  # Regularization

class PF_Trainer(Generic_RHONN_Trainer):
    """Particle Filter Trainer with neuron-specific Q and R values."""
    def __init__(self, n_neurons, n_particles=1000, Q_std=None, R_std=None):
        self.n_neurons = n_neurons
        self.n_particles = n_particles
        
        # Initialize particles
        self.particles = [np.random.randn(n_particles, get_z_size(i)) * 0.1 
                         for i in range(n_neurons)]
        
        # Initialize weights uniformly
        self.weights_pf = [np.ones(n_particles) / n_particles for _ in range(n_neurons)]
        
        # Tuning parameters - neuron-specific or default
        self.Q_std = [Q_std] * n_neurons if isinstance(Q_std, (int, float)) else Q_std or [0.05] * n_neurons
        self.R_std = [R_std] * n_neurons if isinstance(R_std, (int, float)) else R_std or [0.05] * n_neurons
        
        self.regularization_std = 0.001  # Jitter after resampling
        
        # NaN prevention parameters
        self.min_weight = 1e-300
        self.max_log_likelihood = 1000.0
        self.resample_threshold = 0.3
    
    def _normalize_weights(self, weights):
        """Safely normalize weights with NaN and underflow protection."""
        if np.any(~np.isfinite(weights)):
            print("⚠️  Warning: Non-finite weights, resetting to uniform")
            return np.ones_like(weights) / len(weights)
        
        weights = np.maximum(weights, self.min_weight)
        weight_sum = np.sum(weights)
        
        if weight_sum < self.min_weight or not np.isfinite(weight_sum):
            print("⚠️  Warning: Invalid weight sum, resetting to uniform")
            return np.ones_like(weights) / len(weights)
        
        return weights / weight_sum
    
    def _compute_log_likelihood(self, errors, neuron_idx):
        """Compute log-likelihood with numerical stability."""
        errors_clipped = np.clip(errors, -1000, 1000)
        sigma = self.R_std[neuron_idx]
        
        # Student-t log-likelihood with df=3 (heavy tails)
        df = 3.0
        normalized_errors_sq = (errors_clipped / sigma) ** 2
        return -((df + 1) / 2.0) * np.log(1.0 + normalized_errors_sq / df)
    
    def _resample_particles(self, neuron_idx):
        """Systematic resampling with regularization."""
        weights = self.weights_pf[neuron_idx]
        particles = self.particles[neuron_idx]
        n_weights = get_z_size(neuron_idx)
        
        weights = self._normalize_weights(weights)
        
        # Systematic resampling
        positions = (np.arange(self.n_particles) + np.random.random()) / self.n_particles
        cumulative_sum = np.cumsum(weights)
        indices = np.searchsorted(cumulative_sum, positions)
        indices = np.clip(indices, 0, self.n_particles - 1)
        
        # Resample particles
        self.particles[neuron_idx] = particles[indices].copy()
        
        # Add regularization jitter
        jitter = np.random.randn(self.n_particles, n_weights) * self.regularization_std
        self.particles[neuron_idx] += jitter
        
        # Reset weights
        self.weights_pf[neuron_idx] = np.ones(self.n_particles) / self.n_particles
    
    def update(self, x_kp1, x_k, u_k):
        """Update step with NaN prevention and neuron-specific Q/R."""
        for i in range(self.n_neurons):
            try:
                z = construct_z_vector(x_k, u_k, i)
                n_weights = get_z_size(i)
                
                if not np.all(np.isfinite(z)):
                    print(f"⚠️  Warning: Non-finite feature vector for neuron {i}, skipping update")
                    continue
                
                # 1. PREDICTION: Add process noise
                drift = np.random.randn(self.n_particles, n_weights) * self.Q_std[i]
                self.particles[i] += drift
                self.particles[i] = np.clip(self.particles[i], -1000, 1000)
                
                # 2. UPDATE: Compute likelihoods
                preds = self.particles[i] @ z
                
                if not np.all(np.isfinite(preds)):
                    print(f"⚠️  Warning: Non-finite predictions for neuron {i}, resetting particles")
                    self.particles[i] = np.random.randn(self.n_particles, n_weights) * 0.1
                    continue
                
                errors = x_kp1[i] - preds
                log_likelihood = self._compute_log_likelihood(errors, i)
                log_likelihood -= np.max(log_likelihood)  # Shift for stability
                
                # Update weights
                self.weights_pf[i] *= np.exp(log_likelihood)
                self.weights_pf[i] = self._normalize_weights(self.weights_pf[i])
                
                # 3. RESAMPLING: Check effective sample size
                weight_sq_sum = np.sum(self.weights_pf[i] ** 2)
                eff_N = 1.0 / weight_sq_sum if weight_sq_sum > 0 else 0
                
                if eff_N < self.resample_threshold * self.n_particles:
                    self._resample_particles(i)
                    
            except Exception as e:
                print(f"⚠️  Error in PF update for neuron {i}: {e}")
                # Reset to safe state
                self.particles[i] = np.random.randn(self.n_particles, get_z_size(i)) * 0.1
                self.weights_pf[i] = np.ones(self.n_particles) / self.n_particles
    
    def get_estimates(self):
        """Get weighted average of particles."""
        estimates = []
        for i in range(self.n_neurons):
            weights = self._normalize_weights(self.weights_pf[i])
            estimate = np.average(self.particles[i], axis=0, weights=weights)
            
            if not np.all(np.isfinite(estimate)):
                print(f"⚠️  Warning: Non-finite estimate for neuron {i}, using median")
                estimate = np.median(self.particles[i], axis=0)
            
            estimates.append(estimate)
        return estimates

# ============================================================
# 4) ERROR METRICS
# ============================================================

def calculate_error_metrics(y_true, y_pred):
    """Calculate comprehensive error metrics."""
    errors = y_true - y_pred
    
    # Mean Absolute Error (MAE)
    mae = np.mean(np.abs(errors), axis=0)
    mae_total = np.mean(mae)
    
    # Root Mean Square Error (RMSE)
    rmse = np.sqrt(np.mean(errors**2, axis=0))
    rmse_total = np.sqrt(np.mean(rmse**2))
    
    # Normalized RMSE (NRMSE)
    ranges = np.max(y_true, axis=0) - np.min(y_true, axis=0)
    ranges[ranges < 1e-10] = 1.0
    nrmse = rmse / ranges
    nrmse_total = np.mean(nrmse)
    
    # Maximum Absolute Error
    max_error = np.max(np.abs(errors), axis=0)
    
    # R² Score
    ss_res = np.sum(errors**2, axis=0)
    ss_tot = np.sum((y_true - np.mean(y_true, axis=0))**2, axis=0)
    r2 = 1 - (ss_res / (ss_tot + 1e-10))
    r2_total = np.mean(r2)
    
    return {
        'MAE': mae, 'MAE_total': mae_total,
        'RMSE': rmse, 'RMSE_total': rmse_total,
        'NRMSE': nrmse, 'NRMSE_total': nrmse_total,
        'Max_Error': max_error,
        'R2': r2, 'R2_total': r2_total
    }

# ============================================================
# 5) MAIN SIMULATION
# ============================================================

if __name__ == "__main__":
    # Simulation parameters
    n_steps = 1500
    dt = 0.01
    t = np.linspace(0, (n_steps-1)*dt, n_steps)
    n_states = 4  # [θ1, θ2, ω1, ω2]
    
    # Noise parameters
    process_noise_std = [0.001, 0.001, 0.002, 0.002]
    angle_noise_std = 0.01
    omega_noise_std = 0.02
    
    # Initialize weights (uniform distribution)
    np.random.seed(7517)
    initial_weights = []
    for i in range(n_states):
        w_init = np.random.uniform(-1.0, 1.0, size=get_z_size(i))
        initial_weights.append(w_init.copy())
    
    print("Initial RHONN weights:")
    for i, w in enumerate(initial_weights):
        print(f"  Neuron {i}: shape={w.shape}, min={w.min():.4f}, max={w.max():.4f}")
    
    # PF-RHONN neuron-specific parameters
    pf_Q_std = [0.5, 0.5, 0.2, 0.2]          # Process noise
    pf_R_std = [1.0e-1, 1.0e-1, 1.0e-2, 1.0e-2]  # Measurement noise
    
    state_names = ['θ1 (angle 1)', 'θ2 (angle 2)', 'ω1 (ang vel 1)', 'ω2 (ang vel 2)']
    
    print("\nPF-RHONN Neuron-Specific Parameters:")
    for i in range(n_states):
        print(f"  Neuron {i} ({state_names[i]}): Q_std={pf_Q_std[i]:.4f}, R_std={pf_R_std[i]:.4f}")
    
    # Initialize trainers
    ekf = EKF_Trainer(n_states, eta=1.0, P0=1.0, Q=1e-3, R=1e-5)
    ukf = UKF_Trainer(n_states, eta=1.2, alpha=1e-2, Q_std=1e-2, R_std=1e-6)
    pf = PF_Trainer(n_states, n_particles=1100, Q_std=pf_Q_std, R_std=pf_R_std)
    
    # Set initial weights
    for i in range(n_states):
        ekf.weights[i] = initial_weights[i].copy()
        ukf.weights[i] = initial_weights[i].copy()
        pf.particles[i] = np.tile(initial_weights[i], (pf.n_particles, 1))
    
    # Arrays for states
    x_true = np.zeros((n_steps, 4))
    x_est_ekf = np.zeros((n_steps, 4))
    x_est_ukf = np.zeros((n_steps, 4))
    x_est_pf = np.zeros((n_steps, 4))
    y_measured = np.zeros((n_steps, 4))
    
    # Timing arrays
    ekf_training_times, ukf_training_times, pf_training_times = [], [], []
    
    # Initial conditions
    x_true[0] = [np.pi/6, np.pi/4, 0.0, 0.0]
    x_est_ekf[0] = x_true[0]
    x_est_ukf[0] = x_true[0]
    x_est_pf[0] = x_true[0]
    
    # Add Gaussian noise to initial measurement
    initial_noise = np.array([
        np.random.normal(0, angle_noise_std),
        np.random.normal(0, angle_noise_std),
        np.random.normal(0, omega_noise_std),
        np.random.normal(0, omega_noise_std)
    ])
    y_measured[0] = x_true[0] + initial_noise
    
    # Excitation input (external torques)
    u_hist = np.zeros((n_steps, 2))
    for k in range(n_steps):
        τ1 = 0.5 * np.sin(1.0 * t[k]) + 0.3 * np.sin(2.5 * t[k])
        τ2 = 0.4 * np.sin(1.5 * t[k]) + 0.2 * np.cos(3.0 * t[k])
        
        # Add step changes for better excitation
        if 400 < k < 450:
            τ1 += 1.0
            τ2 += 0.5
        elif 900 < k < 950:
            τ1 -= 0.8
            τ2 += 0.6
        elif 1200 < k < 1250:
            τ1 += 0.6
            τ2 -= 0.7
            
        u_hist[k] = [τ1, τ2]

    print("\n" + "="*70)
    print("Simulating Double Pendulum with PARALLEL CONFIGURATION...")
    print("="*70)
    print(f"\nNeuron feature structure:")
    for i in range(n_states):
        print(f"  Neuron {i} ({state_names[i]}): {get_z_size(i)} features")
    print(f"\nSimulation duration: {n_steps*dt:.1f} s ({n_steps} steps)")
    
    # Main simulation loop
    for k in range(n_steps - 1):
        # 1. Generate true next state with process noise
        x_next_clean = plant(x_true[k], u_hist[k], dt)
        process_noise = np.array([np.random.normal(0, std) for std in process_noise_std])
        x_true[k+1] = x_next_clean + process_noise
        
        # 2. Create noisy measurement
        measurement_noise = np.array([
            np.random.normal(0, angle_noise_std),
            np.random.normal(0, angle_noise_std),
            np.random.normal(0, omega_noise_std),
            np.random.normal(0, omega_noise_std)
        ])
        y_measured[k+1] = x_true[k+1] + measurement_noise
        
        # 3. Update filters
        # EKF
        start_time = time.perf_counter()
        ekf.update(y_measured[k+1], x_est_ekf[k], u_hist[k])
        ekf_training_times.append(time.perf_counter() - start_time)
        
        # UKF
        start_time = time.perf_counter()
        ukf.update(y_measured[k+1], x_est_ukf[k], u_hist[k])
        ukf_training_times.append(time.perf_counter() - start_time)
        
        # PF
        start_time = time.perf_counter()
        pf.update(x_true[k+1], x_true[k], u_hist[k])
        pf_training_times.append(time.perf_counter() - start_time)
        
        # Predict next states
        for i in range(4):
            z_ekf = construct_z_vector(y_measured[k], u_hist[k], i)
            x_est_ekf[k+1, i] = np.dot(ekf.weights[i], z_ekf)
            
            z_ukf = construct_z_vector(y_measured[k], u_hist[k], i)
            x_est_ukf[k+1, i] = np.dot(ukf.weights[i], z_ukf)
            
            z_pf = construct_z_vector(y_measured[k], u_hist[k], i)
            w_pf = pf.get_estimates()
            x_est_pf[k+1, i] = np.dot(w_pf[i], z_pf)
        
        # Progress reporting
        if k % 300 == 0 and k > 0:
            print(f"Step {k}/{n_steps-1}")
            err_ekf = np.linalg.norm(x_true[k] - x_est_ekf[k])
            err_ukf = np.linalg.norm(x_true[k] - x_est_ukf[k])
            err_pf = np.linalg.norm(x_true[k] - x_est_pf[k])
            print(f"  Current errors - EKF: {err_ekf:.4f}, UKF: {err_ukf:.4f}, PF: {err_pf:.4f}")

    # ============================================================
    # 6) ERROR ANALYSIS
    # ============================================================
    
    print("\n" + "="*70)
    print("COMPREHENSIVE RESULTS")
    print("="*70)
    
    # Calculate error metrics
    metrics_ekf = calculate_error_metrics(x_true, x_est_ekf)
    metrics_ukf = calculate_error_metrics(x_true, x_est_ukf)
    metrics_pf = calculate_error_metrics(x_true, x_est_pf)
    
    # Print metrics
    for name, metrics in [("EKF-RHONN", metrics_ekf), ("UKF-RHONN", metrics_ukf), ("PF-RHONN", metrics_pf)]:
        print(f"\n--- {name} Performance Metrics ---")
        print(f"RMSE total: {metrics['RMSE_total']:.6f}")
        for i, state_name in enumerate(state_names):
            print(f"  {state_name}: RMSE={metrics['RMSE'][i]:.6f}, MAE={metrics['MAE'][i]:.6f}, "
                  f"R²={metrics['R2'][i]:.4f}")
    
    # Timing analysis
    print("\n" + "="*70)
    print("TRAINING TIMES PER FILTER")
    print("="*70)
    
    filters = ['EKF-RHONN', 'UKF-RHONN', 'PF-RHONN']
    times = [ekf_training_times, ukf_training_times, pf_training_times]
    
    for name, time_list in zip(filters, times):
        mean_time = np.mean(time_list) * 1000  # Convert to ms
        total_time = np.sum(time_list)
        print(f"\n{name}:")
        print(f"  Mean: {mean_time:.4f} ms | Total: {total_time:.6f} s")
    
    # Determine best filter
    rmse_dict = {'EKF-RHONN': metrics_ekf['RMSE_total'], 
                 'UKF-RHONN': metrics_ukf['RMSE_total'], 
                 'PF-RHONN': metrics_pf['RMSE_total']}
    best_filter = min(rmse_dict, key=rmse_dict.get)
    
    print("\n" + "="*70)
    print(f"BEST FILTER: {best_filter} (RMSE total: {rmse_dict[best_filter]:.6f})")
    print("="*70)

# ============================================================
# 7) VISUALIZATION
# ============================================================

print("\nGenerating visualizations...")

# Configuration for thesis-style plots
thesis_config = {
    'font_family': 'Computer Modern, serif',
    'font_size': 16,
    'title_font_size': 18,
    'legend_font_size': 14,
    'line_width_est': 2.0,
    'plot_width': 1000,
    'plot_height': 500,
    'grid_color': 'rgba(200, 200, 200, 0.3)',
    'grid_width': 0.5
}

# --- Training Times Plot ---
fig_times = go.Figure()
sim_time = t[1:]  # Exclude first time step

fig_times.add_trace(go.Scatter(
    x=sim_time, y=pf_training_times,
    mode='lines',
    name='PF-RHONN',
    line=dict(color='#d62728', width=2),
))

fig_times.update_layout(
    title={'text': 'Training Time per Iteration', 'x': 0.5},
    xaxis_title='Simulation Time (s)',
    yaxis_title='Training Time (s)',
    xaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color'], 
               gridwidth=thesis_config['grid_width'], showline=True),
    yaxis=dict(type='log', showgrid=True, gridcolor=thesis_config['grid_color'],
               gridwidth=thesis_config['grid_width'], showline=True),
    legend=dict(x=0.02, y=0.98, bgcolor='rgba(255, 255, 255, 0.9)'),
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    width=thesis_config['plot_width'],
    height=thesis_config['plot_height']
)

fig_times.show()

# --- Average Training Time Bar Chart ---
fig_times_bar = go.Figure()

filters = ['EKF-RHONN', 'UKF-RHONN', 'PF-RHONN']
mean_times_ms = [np.mean(ekf_training_times)*1000, 
                 np.mean(ukf_training_times)*1000, 
                 np.mean(pf_training_times)*1000]

fig_times_bar.add_trace(go.Bar(
    x=filters,
    y=mean_times_ms,
    marker_color=['#1f77b4', '#2ca02c', '#d62728'],
    text=[f'{t:.3f} ms' for t in mean_times_ms],
    textposition='outside'
))

fig_times_bar.update_layout(
    title={'text': 'Average Training Time per Iteration', 'x': 0.5},
    xaxis_title='Filter Type',
    yaxis_title='Average Time (ms)',
    yaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color']),
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    width=thesis_config['plot_width'],
    height=thesis_config['plot_height']
)

fig_times_bar.show()

# --- State Plots ---
states_info = [
    {'idx': 0, 'var': 'θ₁', 'desc': 'First Pendulum Angle', 'y_label': 'Angle θ₁ (rad)'},
    {'idx': 1, 'var': 'θ₂', 'desc': 'Second Pendulum Angle', 'y_label': 'Angle θ₂ (rad)'},
    {'idx': 2, 'var': 'ω₁', 'desc': 'Angular Velocity 1', 'y_label': 'Velocity ω₁ (rad/s)'},
    {'idx': 3, 'var': 'ω₂', 'desc': 'Angular Velocity 2', 'y_label': 'Velocity ω₂ (rad/s)'}
]

for state_info in states_info:
    i = state_info['idx']
    
    fig = go.Figure()
    
    # Measured state (with noise)
    fig.add_trace(go.Scatter(
        x=t, y=y_measured[:, i],
        mode='lines',
        name='Measurement (with noise)',
        line=dict(color='#808080', width=thesis_config['line_width_est']),
        showlegend=True
    ))
    
    # Filter estimates
    fig.add_trace(go.Scatter(
        x=t, y=x_est_ekf[:, i], mode='lines',
        name='EKF-RHONN', line=dict(color='#1f77b4', width=thesis_config['line_width_est'], dash='dash')
    ))
    
    fig.add_trace(go.Scatter(
        x=t, y=x_est_ukf[:, i], mode='lines',
        name='UKF-RHONN', line=dict(color='#2ca02c', width=thesis_config['line_width_est'], dash='dot')
    ))
    
    fig.add_trace(go.Scatter(
        x=t, y=x_est_pf[:, i], mode='lines',
        name='PF-RHONN', line=dict(color='#d62728', width=thesis_config['line_width_est'], dash='dashdot')
    ))
    
    fig.update_layout(
        title={'text': f'State {state_info["var"]}: {state_info["desc"]}', 'x': 0.5},
        xaxis_title='Time (s)',
        yaxis_title=state_info['y_label'],
        xaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color']),
        yaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color']),
        legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='center', x=0.5),
        font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
        plot_bgcolor='white',
        width=thesis_config['plot_width'],
        height=thesis_config['plot_height']
    )
    
    fig.show()

# --- Metrics Comparison Bar Chart ---
fig_metrics = go.Figure()

metrics_names = ['RMSE', 'MAE', 'NRMSE']
ekf_vals = [metrics_ekf['RMSE_total'], metrics_ekf['MAE_total'], metrics_ekf['NRMSE_total']]
ukf_vals = [metrics_ukf['RMSE_total'], metrics_ukf['MAE_total'], metrics_ukf['NRMSE_total']]
pf_vals = [metrics_pf['RMSE_total'], metrics_pf['MAE_total'], metrics_pf['NRMSE_total']]

x = np.arange(len(metrics_names))
width = 0.25

fig_metrics.add_trace(go.Bar(
    x=x - width, y=ekf_vals, width=width,
    name='EKF-RHONN', marker_color='#1f77b4'
))

fig_metrics.add_trace(go.Bar(
    x=x, y=ukf_vals, width=width,
    name='UKF-RHONN', marker_color='#2ca02c'
))

fig_metrics.add_trace(go.Bar(
    x=x + width, y=pf_vals, width=width,
    name='PF-RHONN', marker_color='#d62728'
))

fig_metrics.update_layout(
    title={'text': 'Error Metrics Comparison', 'x': 0.5},
    xaxis=dict(tickmode='array', tickvals=x, ticktext=metrics_names, showline=True),
    yaxis=dict(title='Metric Value', showgrid=True, gridcolor=thesis_config['grid_color']),
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='center', x=0.5),
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    width=thesis_config['plot_width'],
    height=thesis_config['plot_height'],
    barmode='group'
)

fig_metrics.show()

print("\n✅ Simulation completed successfully.")
```

```python
# Neural Identifier Training for Double Pendulum System
# Training Methods: EKF, UKF, Particle Filter (with neuron-specific parameters)
# WITH NON-GAUSSIAN NOISE (Impulsive + Mixture + Heavy-tailed)

import numpy as np
import plotly.graph_objects as go
import warnings
import time

warnings.filterwarnings('ignore')

# ============================================================
# 1) TRUE SYSTEM: Double Pendulum Dynamics
# ============================================================

def plant_dynamics(state, u):
    """
    Dynamics of a double pendulum with damping and external torques.
    state = [θ1, θ2, ω1, ω2] 
    u = [τ1, τ2]: external torques on joints
    """
    # Physical Parameters
    m1, m2 = 1.0, 1.0      # Masses (kg)
    l1, l2 = 1.0, 1.0      # Lengths (m)
    g = 9.81                # Gravity (m/s²)
    b1, b2 = 0.1, 0.1      # Damping coefficients (N⋅m⋅s)
    
    θ1, θ2, ω1, ω2 = state
    τ1, τ2 = u
    
    # Precompute trigonometric terms
    c12 = np.cos(θ1 - θ2)
    s12 = np.sin(θ1 - θ2)
    s1, s2 = np.sin(θ1), np.sin(θ2)
    
    # Mass matrix elements
    M11 = (m1 + m2) * l1**2
    M12 = m2 * l1 * l2 * c12
    M21 = M12
    M22 = m2 * l2**2
    
    # Coriolis and centrifugal terms
    h = m2 * l1 * l2 * ω2**2 * s12
    C1 = h - (m1 + m2) * g * l1 * s1 - b1 * ω1 + τ1
    C2 = -m2 * l1 * l2 * ω1**2 * s12 - m2 * g * l2 * s2 - b2 * ω2 + τ2
    
    # Solve for accelerations
    det_M = M11 * M22 - M12 * M21
    if abs(det_M) < 1e-10:
        det_M = 1e-10  # Prevent division by zero
    
    α1 = (M22 * C1 - M12 * C2) / det_M
    α2 = (-M21 * C1 + M11 * C2) / det_M
    
    # State derivatives
    return np.array([ω1, ω2, α1, α2])

def plant(x_k, u_k, dt=0.01):
    """RK4 integration step."""
    k1 = plant_dynamics(x_k, u_k)
    k2 = plant_dynamics(x_k + 0.5*dt*k1, u_k)
    k3 = plant_dynamics(x_k + 0.5*dt*k2, u_k)
    k4 = plant_dynamics(x_k + dt*k3, u_k)
    
    return x_k + (dt/6.0) * (k1 + 2*k2 + 2*k3 + k4)

# ============================================================
# 2) NON-GAUSSIAN NOISE GENERATORS
# ============================================================

class NonGaussianNoise:
    """Collection of non-Gaussian noise generators."""
    
    @staticmethod
    def laplace_noise(loc=0.0, scale=1.0, size=1):
        """Laplace (double exponential) noise - heavier tails than Gaussian."""
        u = np.random.uniform(-0.5, 0.5, size=size)
        return loc - scale * np.sign(u) * np.log(1 - 2 * np.abs(u))
    
    @staticmethod
    def student_t_noise(df=3.0, scale=1.0, size=1):
        """Student's t-distribution noise - heavy tails."""
        return np.random.standard_t(df, size=size) * scale
    
    @staticmethod
    def cauchy_noise(loc=0.0, scale=1.0, size=1):
        """Cauchy noise - very heavy tails, no finite variance."""
        return np.random.standard_cauchy(size=size) * scale + loc
    
    @staticmethod
    def uniform_noise(low=-1.0, high=1.0, size=1):
        """Uniform noise - bounded, non-Gaussian."""
        return np.random.uniform(low, high, size=size)
    
    @staticmethod
    def gaussian_mixture_noise(means=[0, 0], stds=[1, 5], weights=[0.9, 0.1], size=1):
        """Gaussian mixture noise - good for impulsive noise."""
        choices = np.random.choice(len(weights), size=size, p=weights)
        noise = np.zeros(size)
        for i, (mean, std) in enumerate(zip(means, stds)):
            mask = (choices == i)
            noise[mask] = np.random.normal(mean, std, size=np.sum(mask))
        return noise
    
    @staticmethod
    def impulsive_noise(std_normal=1.0, std_impulse=10.0, impulse_prob=0.05, size=1):
        """Impulsive noise with rare large outliers."""
        impulse_mask = np.random.random(size) < impulse_prob
        noise = np.random.normal(0, std_normal, size)
        noise[impulse_mask] = np.random.normal(0, std_impulse, size=np.sum(impulse_mask))
        return noise
    
    @staticmethod
    def exponential_noise(scale=1.0, size=1, symmetric=False):
        """Exponential noise (asymmetric)."""
        noise = np.random.exponential(scale, size)
        if symmetric:
            noise = noise * np.sign(np.random.uniform(-1, 1, size))
        return noise

# ============================================================
# 3) RHONN STRUCTURE - Neuron-specific features
# ============================================================

def sigmoidal(z, beta=1.0):
    """Sigmoid activation function."""
    z = np.clip(z, -50, 50) 
    return 1.0 / (1.0 + np.exp(-beta * z))

def construct_z_vector(x_est, u_input, neuron_index):
    """
    Features for Double Pendulum - specific for each neuron.
    x_est = [θ1, θ2, ω1, ω2]
    u_input = [τ1, τ2]
    neuron_index: 0=θ1, 1=θ2, 2=ω1, 3=ω2
    """
    θ1, θ2, ω1, ω2 = x_est
    τ1, τ2 = u_input
    
    # Sigmoid terms
    s_θ1 = sigmoidal(θ1)
    s_θ2 = sigmoidal(θ2)
    s_ω1 = sigmoidal(ω1)
    s_ω2 = sigmoidal(ω2)
    s_τ1 = sigmoidal(τ1)
    s_τ2 = sigmoidal(τ2)
    
    # Neuron-specific features
    if neuron_index == 0:  # θ1 (angle of first pendulum)
        return np.array([s_ω1**3, s_θ1**2, s_θ1 * s_ω1])
    
    elif neuron_index == 1:  # θ2 (angle of second pendulum)
        return np.array([s_ω2, s_ω2**2, s_θ2 * s_ω2, s_ω1])
    
    elif neuron_index == 2:  # ω1 (angular velocity of first pendulum)
        return np.array([s_ω1, s_ω1**2, s_τ1, s_ω1 * s_ω2])
    
    elif neuron_index == 3:  # ω2 (angular velocity of second pendulum)
        return np.array([s_ω2, s_ω2**2, s_τ2, s_ω1 * s_ω2])
    
    else:
        raise ValueError(f"Invalid neuron index: {neuron_index}")

def get_z_size(neuron_index):
    """Return number of features for a given neuron."""
    sizes = [3, 4, 4, 4]  # [θ1, θ2, ω1, ω2]
    return sizes[neuron_index]

# ============================================================
# 4) TRAINERS (EKF, UKF, PF) - Adapted for multiple sizes
# ============================================================

class Generic_RHONN_Trainer:
    """Base class for training loop logic."""
    def get_prediction(self, weights, x_k, u_k, neuron_idx):
        z = construct_z_vector(x_k, u_k, neuron_idx)
        return np.dot(weights, z)

class EKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, eta=1.0, P0=1.0, Q=1e-3, R=1e-5):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(get_z_size(i))*0.1 for i in range(n_neurons)]
        self.P = [np.eye(get_z_size(i))*P0 for i in range(n_neurons)]
        self.Q_matrices = [np.eye(get_z_size(i))*Q for i in range(n_neurons)]
        self.R = R
        self.eta = eta

    def update(self, x_kp1, x_k, u_k):
        for i in range(self.n_neurons):
            z = construct_z_vector(x_k, u_k, i)
            H_i = z.reshape(-1, 1)
            
            x_hat_i = np.dot(self.weights[i], z)
            e_i = x_kp1[i] - x_hat_i
            
            # Kalman Gain
            S = self.R + (H_i.T @ self.P[i] @ H_i)[0, 0]
            K_i = (self.P[i] @ H_i).flatten() / S
            
            # Weight update
            self.weights[i] += self.eta * K_i * e_i
            
            # Covariance update
            self.P[i] = self.P[i] - np.outer(K_i, H_i.flatten()) @ self.P[i] + self.Q_matrices[i]

class UKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, eta=1.0, alpha=1e-2, Q_std=1e-3, R_std=1e-5):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(get_z_size(i))*0.1 for i in range(n_neurons)]
        self.P = [np.eye(get_z_size(i)) for i in range(n_neurons)]
        self.Q_matrices = [np.eye(get_z_size(i))*Q_std for i in range(n_neurons)]
        self.R = R_std
        self.eta = eta
        self.alpha = alpha

    def update(self, x_kp1, x_k, u_k):
        for i in range(self.n_neurons):
            z = construct_z_vector(x_k, u_k, i)
            n = get_z_size(i)
            
            # Sigma points parameters
            lambda_ = self.alpha**2 * n - n
            Wm = np.full(2*n+1, 1/(2*(n+lambda_)))
            Wc = np.copy(Wm)
            Wm[0] = lambda_/(n+lambda_)
            Wc[0] = Wm[0] + (3 - self.alpha**2)
            
            # Generate sigma points
            try:
                L = np.linalg.cholesky((n + lambda_) * self.P[i])
            except:
                L = np.eye(n) * 0.1
                
            sigmas = np.zeros((2*n+1, n))
            sigmas[0] = self.weights[i]
            for k in range(n):
                sigmas[k+1] = self.weights[i] + L[:,k]
                sigmas[n+k+1] = self.weights[i] - L[:,k]
            
            # Transform through measurement function
            Y_sigmas = np.dot(sigmas, z)
            y_mean = np.sum(Wm * Y_sigmas)
            
            # Covariances
            Py = np.sum(Wc * (Y_sigmas - y_mean)**2) + self.R
            Pxy = np.zeros(n)
            for k in range(2*n+1):
                Pxy += Wc[k] * (sigmas[k] - self.weights[i]) * (Y_sigmas[k] - y_mean)
                
            # Update
            K = Pxy / Py
            err = x_kp1[i] - y_mean
            self.weights[i] += self.eta * K * err
            self.P[i] -= np.outer(K, K) * Py
            self.P[i] += np.eye(n)*1e-6  # Regularization

class PF_Trainer(Generic_RHONN_Trainer):
    """Particle Filter Trainer with neuron-specific Q and R values."""
    def __init__(self, n_neurons, n_particles=1000, Q_std=None, R_std=None):
        self.n_neurons = n_neurons
        self.n_particles = n_particles
        
        # Initialize particles
        self.particles = [np.random.randn(n_particles, get_z_size(i)) * 0.1 
                         for i in range(n_neurons)]
        
        # Initialize weights uniformly
        self.weights_pf = [np.ones(n_particles) / n_particles for _ in range(n_neurons)]
        
        # Tuning parameters - neuron-specific or default
        self.Q_std = [Q_std] * n_neurons if isinstance(Q_std, (int, float)) else Q_std or [0.05] * n_neurons
        self.R_std = [R_std] * n_neurons if isinstance(R_std, (int, float)) else R_std or [0.05] * n_neurons
        
        self.regularization_std = 0.001  # Jitter after resampling
        
        # NaN prevention parameters
        self.min_weight = 1e-300
        self.max_log_likelihood = 1000.0
        self.resample_threshold = 0.3
    
    def _normalize_weights(self, weights):
        """Safely normalize weights with NaN and underflow protection."""
        if np.any(~np.isfinite(weights)):
            print("⚠️  Warning: Non-finite weights, resetting to uniform")
            return np.ones_like(weights) / len(weights)
        
        weights = np.maximum(weights, self.min_weight)
        weight_sum = np.sum(weights)
        
        if weight_sum < self.min_weight or not np.isfinite(weight_sum):
            print("⚠️  Warning: Invalid weight sum, resetting to uniform")
            return np.ones_like(weights) / len(weights)
        
        return weights / weight_sum
    
    def _compute_log_likelihood(self, errors, neuron_idx):
        """Compute log-likelihood with numerical stability using Student-t."""
        errors_clipped = np.clip(errors, -1000, 1000)
        sigma = self.R_std[neuron_idx]
        
        # Student-t log-likelihood with df=3 (heavy tails - robust to non-Gaussian)
        df = 3.0
        normalized_errors_sq = (errors_clipped / sigma) ** 2
        return -((df + 1) / 2.0) * np.log(1.0 + normalized_errors_sq / df)
    
    def _resample_particles(self, neuron_idx):
        """Systematic resampling with regularization."""
        weights = self.weights_pf[neuron_idx]
        particles = self.particles[neuron_idx]
        n_weights = get_z_size(neuron_idx)
        
        weights = self._normalize_weights(weights)
        
        # Systematic resampling
        positions = (np.arange(self.n_particles) + np.random.random()) / self.n_particles
        cumulative_sum = np.cumsum(weights)
        indices = np.searchsorted(cumulative_sum, positions)
        indices = np.clip(indices, 0, self.n_particles - 1)
        
        # Resample particles
        self.particles[neuron_idx] = particles[indices].copy()
        
        # Add regularization jitter
        jitter = np.random.randn(self.n_particles, n_weights) * self.regularization_std
        self.particles[neuron_idx] += jitter
        
        # Reset weights
        self.weights_pf[neuron_idx] = np.ones(self.n_particles) / self.n_particles
    
    def update(self, x_kp1, x_k, u_k):
        """Update step with NaN prevention and neuron-specific Q/R."""
        for i in range(self.n_neurons):
            try:
                z = construct_z_vector(x_k, u_k, i)
                n_weights = get_z_size(i)
                
                if not np.all(np.isfinite(z)):
                    print(f"⚠️  Warning: Non-finite feature vector for neuron {i}, skipping update")
                    continue
                
                # 1. PREDICTION: Add process noise
                drift = np.random.randn(self.n_particles, n_weights) * self.Q_std[i]
                self.particles[i] += drift
                self.particles[i] = np.clip(self.particles[i], -1000, 1000)
                
                # 2. UPDATE: Compute likelihoods
                preds = self.particles[i] @ z
                
                if not np.all(np.isfinite(preds)):
                    print(f"⚠️  Warning: Non-finite predictions for neuron {i}, resetting particles")
                    self.particles[i] = np.random.randn(self.n_particles, n_weights) * 0.1
                    continue
                
                errors = x_kp1[i] - preds
                log_likelihood = self._compute_log_likelihood(errors, i)
                log_likelihood -= np.max(log_likelihood)  # Shift for stability
                
                # Update weights
                self.weights_pf[i] *= np.exp(log_likelihood)
                self.weights_pf[i] = self._normalize_weights(self.weights_pf[i])
                
                # 3. RESAMPLING: Check effective sample size
                weight_sq_sum = np.sum(self.weights_pf[i] ** 2)
                eff_N = 1.0 / weight_sq_sum if weight_sq_sum > 0 else 0
                
                if eff_N < self.resample_threshold * self.n_particles:
                    self._resample_particles(i)
                    
            except Exception as e:
                print(f"⚠️  Error in PF update for neuron {i}: {e}")
                # Reset to safe state
                self.particles[i] = np.random.randn(self.n_particles, get_z_size(i)) * 0.1
                self.weights_pf[i] = np.ones(self.n_particles) / self.n_particles
    
    def get_estimates(self):
        """Get weighted average of particles."""
        estimates = []
        for i in range(self.n_neurons):
            weights = self._normalize_weights(self.weights_pf[i])
            estimate = np.average(self.particles[i], axis=0, weights=weights)
            
            if not np.all(np.isfinite(estimate)):
                print(f"⚠️  Warning: Non-finite estimate for neuron {i}, using median")
                estimate = np.median(self.particles[i], axis=0)
            
            estimates.append(estimate)
        return estimates

# ============================================================
# 5) ERROR METRICS
# ============================================================

def calculate_error_metrics(y_true, y_pred):
    """Calculate comprehensive error metrics."""
    errors = y_true - y_pred
    
    # Mean Absolute Error (MAE)
    mae = np.mean(np.abs(errors), axis=0)
    mae_total = np.mean(mae)
    
    # Root Mean Square Error (RMSE)
    rmse = np.sqrt(np.mean(errors**2, axis=0))
    rmse_total = np.sqrt(np.mean(rmse**2))
    
    # Normalized RMSE (NRMSE)
    ranges = np.max(y_true, axis=0) - np.min(y_true, axis=0)
    ranges[ranges < 1e-10] = 1.0
    nrmse = rmse / ranges
    nrmse_total = np.mean(nrmse)
    
    # Maximum Absolute Error
    max_error = np.max(np.abs(errors), axis=0)
    
    # R² Score
    ss_res = np.sum(errors**2, axis=0)
    ss_tot = np.sum((y_true - np.mean(y_true, axis=0))**2, axis=0)
    r2 = 1 - (ss_res / (ss_tot + 1e-10))
    r2_total = np.mean(r2)
    
    return {
        'MAE': mae, 'MAE_total': mae_total,
        'RMSE': rmse, 'RMSE_total': rmse_total,
        'NRMSE': nrmse, 'NRMSE_total': nrmse_total,
        'Max_Error': max_error,
        'R2': r2, 'R2_total': r2_total
    }

# ============================================================
# 6) MAIN SIMULATION WITH NON-GAUSSIAN NOISE
# ============================================================

if __name__ == "__main__":
    # Simulation parameters
    n_steps = 1500
    dt = 0.01
    t = np.linspace(0, (n_steps-1)*dt, n_steps)
    n_states = 4  # [θ1, θ2, ω1, ω2]
    
    # NON-GAUSSIAN NOISE PARAMETERS
    print("\n" + "="*70)
    print("NON-GAUSSIAN NOISE CONFIGURATION")
    print("="*70)
    print("Process Noise Types:")
    print("  - θ1, θ2: Laplace noise (heavy-tailed)")
    print("  - ω1, ω2: Gaussian mixture (impulsive)")
    print("\nMeasurement Noise Types:")
    print("  - θ1, θ2: Student's t-distribution (df=3, heavy-tailed)")
    print("  - ω1, ω2: Impulsive noise (5% outliers)")
    
    # Process noise parameters (non-Gaussian)
    process_noise_params = [
        {'type': 'laplace', 'scale': 0.002},    # θ1: Laplace
        {'type': 'laplace', 'scale': 0.002},    # θ2: Laplace  
        {'type': 'mixture', 'std_normal': 0.001, 'std_impulse': 0.01, 'impulse_prob': 0.1},  # ω1: Mixture
        {'type': 'mixture', 'std_normal': 0.001, 'std_impulse': 0.01, 'impulse_prob': 0.1}   # ω2: Mixture
    ]
    
    # Measurement noise parameters (non-Gaussian)
    measurement_noise_params = [
        {'type': 'student_t', 'df': 3, 'scale': 0.015},  # θ1: Student's t
        {'type': 'student_t', 'df': 3, 'scale': 0.015},  # θ2: Student's t
        {'type': 'impulsive', 'std_normal': 0.02, 'std_impulse': 0.2, 'impulse_prob': 0.05},  # ω1: Impulsive
        {'type': 'impulsive', 'std_normal': 0.02, 'std_impulse': 0.2, 'impulse_prob': 0.05}   # ω2: Impulsive
    ]
    
    # Function to generate non-Gaussian noise
    def generate_non_gaussian_noise(params, size=1):
        """Generate non-Gaussian noise based on parameters."""
        noise_type = params['type']
        
        if noise_type == 'laplace':
            return NonGaussianNoise.laplace_noise(scale=params['scale'], size=size)
        elif noise_type == 'student_t':
            return NonGaussianNoise.student_t_noise(df=params['df'], scale=params['scale'], size=size)
        elif noise_type == 'mixture':
            return NonGaussianNoise.gaussian_mixture_noise(
                means=[0, 0], 
                stds=[params['std_normal'], params['std_impulse']], 
                weights=[1-params['impulse_prob'], params['impulse_prob']], 
                size=size
            )
        elif noise_type == 'impulsive':
            return NonGaussianNoise.impulsive_noise(
                std_normal=params['std_normal'],
                std_impulse=params['std_impulse'],
                impulse_prob=params['impulse_prob'],
                size=size
            )
        elif noise_type == 'cauchy':
            return NonGaussianNoise.cauchy_noise(scale=params['scale'], size=size)
        elif noise_type == 'uniform':
            return NonGaussianNoise.uniform_noise(
                low=-params['range'], 
                high=params['range'], 
                size=size
            )
        else:
            # Fallback to Gaussian
            return np.random.normal(0, params.get('std', 0.01), size)
    
    # Initialize weights (uniform distribution)
    np.random.seed(7517)
    initial_weights = []
    for i in range(n_states):
        w_init = np.random.uniform(-1.0, 1.0, size=get_z_size(i))
        initial_weights.append(w_init.copy())
    
    print("\nInitial RHONN weights:")
    for i, w in enumerate(initial_weights):
        print(f"  Neuron {i}: shape={w.shape}, min={w.min():.4f}, max={w.max():.4f}")
    
    # PF-RHONN neuron-specific parameters (tuned for non-Gaussian noise)
    pf_Q_std = [0.5, 0.5, 0.5, 0.5]          # Higher Q for non-Gaussian process noise
    pf_R_std = [1e-3, 1e-3, 1e-3, 1e-3]  # Higher R for non-Gaussian measurement noise
    
    state_names = ['θ1 (angle 1)', 'θ2 (angle 2)', 'ω1 (ang vel 1)', 'ω2 (ang vel 2)']
    
    print("\nPF-RHONN Neuron-Specific Parameters (adjusted for non-Gaussian noise):")
    for i in range(n_states):
        print(f"  Neuron {i} ({state_names[i]}): Q_std={pf_Q_std[i]:.4f}, R_std={pf_R_std[i]:.4f}")
    
    # Initialize trainers (with increased Q/R for non-Gaussian noise)
    ekf = EKF_Trainer(n_states, eta=1.0, P0=1.0, Q=5e-3, R=5e-4)  # Increased Q/R
    ukf = UKF_Trainer(n_states, eta=1.2, alpha=1e-2, Q_std=5e-2, R_std=1e-4)  # Increased Q_std/R_std
    pf = PF_Trainer(n_states, n_particles=1100, Q_std=pf_Q_std, R_std=pf_R_std)
    
    # Set initial weights
    for i in range(n_states):
        ekf.weights[i] = initial_weights[i].copy()
        ukf.weights[i] = initial_weights[i].copy()
        pf.particles[i] = np.tile(initial_weights[i], (pf.n_particles, 1))
    
    # Arrays for states
    x_true = np.zeros((n_steps, 4))
    x_est_ekf = np.zeros((n_steps, 4))
    x_est_ukf = np.zeros((n_steps, 4))
    x_est_pf = np.zeros((n_steps, 4))
    y_measured = np.zeros((n_steps, 4))
    
    # Timing arrays
    ekf_training_times, ukf_training_times, pf_training_times = [], [], []
    
    # Initial conditions
    x_true[0] = [np.pi/6, np.pi/4, 0.0, 0.0]
    x_est_ekf[0] = x_true[0]
    x_est_ukf[0] = x_true[0]
    x_est_pf[0] = x_true[0]
    
    # Add non-Gaussian noise to initial measurement
    initial_noise = np.array([
        generate_non_gaussian_noise(measurement_noise_params[0], 1)[0],
        generate_non_gaussian_noise(measurement_noise_params[1], 1)[0],
        generate_non_gaussian_noise(measurement_noise_params[2], 1)[0],
        generate_non_gaussian_noise(measurement_noise_params[3], 1)[0]
    ])
    y_measured[0] = x_true[0] + initial_noise
    
    # Excitation input (external torques)
    u_hist = np.zeros((n_steps, 2))
    for k in range(n_steps):
        τ1 = 0.5 * np.sin(1.0 * t[k]) + 0.3 * np.sin(2.5 * t[k])
        τ2 = 0.4 * np.sin(1.5 * t[k]) + 0.2 * np.cos(3.0 * t[k])
        
        # Add step changes for better excitation
        if 400 < k < 450:
            τ1 += 1.0
            τ2 += 0.5
        elif 900 < k < 950:
            τ1 -= 0.8
            τ2 += 0.6
        elif 1200 < k < 1250:
            τ1 += 0.6
            τ2 -= 0.7
            
        u_hist[k] = [τ1, τ2]
    
    # Arrays to track noise statistics
    process_noise_history = np.zeros((n_steps-1, 4))
    measurement_noise_history = np.zeros((n_steps, 4))
    
    print("\n" + "="*70)
    print("Simulating Double Pendulum with NON-GAUSSIAN NOISE")
    print("="*70)
    print(f"\nNeuron feature structure:")
    for i in range(n_states):
        print(f"  Neuron {i} ({state_names[i]}): {get_z_size(i)} features")
    print(f"\nSimulation duration: {n_steps*dt:.1f} s ({n_steps} steps)")
    
    # Main simulation loop with NON-GAUSSIAN NOISE
    for k in range(n_steps - 1):
        # 1. Generate true next state with NON-GAUSSIAN process noise
        x_next_clean = plant(x_true[k], u_hist[k], dt)
        
        # Add NON-GAUSSIAN process noise
        process_noise = np.array([
            generate_non_gaussian_noise(process_noise_params[0], 1)[0],
            generate_non_gaussian_noise(process_noise_params[1], 1)[0],
            generate_non_gaussian_noise(process_noise_params[2], 1)[0],
            generate_non_gaussian_noise(process_noise_params[3], 1)[0]
        ])
        
        process_noise_history[k] = process_noise
        x_true[k+1] = x_next_clean + process_noise
        
        # 2. Create NON-GAUSSIAN noisy measurement
        measurement_noise = np.array([
            generate_non_gaussian_noise(measurement_noise_params[0], 1)[0],
            generate_non_gaussian_noise(measurement_noise_params[1], 1)[0],
            generate_non_gaussian_noise(measurement_noise_params[2], 1)[0],
            generate_non_gaussian_noise(measurement_noise_params[3], 1)[0]
        ])
        
        measurement_noise_history[k+1] = measurement_noise
        y_measured[k+1] = x_true[k+1] + measurement_noise
        
        # 3. Update filters (filters still assume Gaussian noise!)
        # EKF
        start_time = time.perf_counter()
        ekf.update(y_measured[k+1], x_est_ekf[k], u_hist[k])
        ekf_training_times.append(time.perf_counter() - start_time)
        
        # UKF
        start_time = time.perf_counter()
        ukf.update(y_measured[k+1], x_est_ukf[k], u_hist[k])
        ukf_training_times.append(time.perf_counter() - start_time)
        
        # PF
        start_time = time.perf_counter()
        pf.update(x_true[k+1], x_true[k], u_hist[k])
        pf_training_times.append(time.perf_counter() - start_time)
        
        # Predict next states
        for i in range(4):
            z_ekf = construct_z_vector(y_measured[k], u_hist[k], i)
            x_est_ekf[k+1, i] = np.dot(ekf.weights[i], z_ekf)
            
            z_ukf = construct_z_vector(y_measured[k], u_hist[k], i)
            x_est_ukf[k+1, i] = np.dot(ukf.weights[i], z_ukf)
            
            z_pf = construct_z_vector(y_measured[k], u_hist[k], i)
            w_pf = pf.get_estimates()
            x_est_pf[k+1, i] = np.dot(w_pf[i], z_pf)
        
        # Progress reporting
        if k % 300 == 0 and k > 0:
            print(f"Step {k}/{n_steps-1}")
            err_ekf = np.linalg.norm(x_true[k] - x_est_ekf[k])
            err_ukf = np.linalg.norm(x_true[k] - x_est_ukf[k])
            err_pf = np.linalg.norm(x_true[k] - x_est_pf[k])
            print(f"  Current errors - EKF: {err_ekf:.4f}, UKF: {err_ukf:.4f}, PF: {err_pf:.4f}")
            
            # Show some noise statistics
            if k == 300:
                print("\n  Non-Gaussian Noise Statistics (first 300 steps):")
                print("  Process noise (abs max):", np.max(np.abs(process_noise_history[:k]), axis=0))
                print("  Measurement noise (abs max):", np.max(np.abs(measurement_noise_history[:k]), axis=0))

    # ============================================================
    # 7) ERROR ANALYSIS WITH NON-GAUSSIAN NOISE
    # ============================================================
    
    print("\n" + "="*70)
    print("COMPREHENSIVE RESULTS WITH NON-GAUSSIAN NOISE")
    print("="*70)
    
    # Calculate error metrics
    metrics_ekf = calculate_error_metrics(x_true, x_est_ekf)
    metrics_ukf = calculate_error_metrics(x_true, x_est_ukf)
    metrics_pf = calculate_error_metrics(x_true, x_est_pf)
    
    # Print metrics
    for name, metrics in [("EKF-RHONN", metrics_ekf), ("UKF-RHONN", metrics_ukf), ("PF-RHONN", metrics_pf)]:
        print(f"\n--- {name} Performance Metrics ---")
        print(f"RMSE total: {metrics['RMSE_total']:.6f}")
        for i, state_name in enumerate(state_names):
            print(f"  {state_name}: RMSE={metrics['RMSE'][i]:.6f}, MAE={metrics['MAE'][i]:.6f}, "
                  f"R²={metrics['R2'][i]:.4f}")
    
    # Analyze noise characteristics
    print("\n" + "="*70)
    print("NON-GAUSSIAN NOISE STATISTICS")
    print("="*70)
    
    print("\nProcess Noise Statistics:")
    for i in range(4):
        noise = process_noise_history[:, i]
        print(f"\n  {state_names[i]}:")
        print(f"    Std: {np.std(noise):.6f}")
        print(f"    Mean: {np.mean(noise):.6f}")
        print(f"    Kurtosis: {np.mean((noise - np.mean(noise))**4) / np.std(noise)**4:.4f} (Gaussian=3)")
        print(f"    Max abs: {np.max(np.abs(noise)):.6f}")
        print(f"    |Noise| > 3σ: {np.sum(np.abs(noise) > 3*np.std(noise))} samples")
    
    print("\nMeasurement Noise Statistics:")
    for i in range(4):
        noise = measurement_noise_history[:, i]
        print(f"\n  {state_names[i]}:")
        print(f"    Std: {np.std(noise):.6f}")
        print(f"    Mean: {np.mean(noise):.6f}")
        print(f"    Kurtosis: {np.mean((noise - np.mean(noise))**4) / np.std(noise)**4:.4f} (Gaussian=3)")
        print(f"    Max abs: {np.max(np.abs(noise)):.6f}")
        print(f"    |Noise| > 3σ: {np.sum(np.abs(noise) > 3*np.std(noise))} samples")
    
    # Timing analysis
    print("\n" + "="*70)
    print("TRAINING TIMES PER FILTER")
    print("="*70)
    
    filters = ['EKF-RHONN', 'UKF-RHONN', 'PF-RHONN']
    times = [ekf_training_times, ukf_training_times, pf_training_times]
    
    for name, time_list in zip(filters, times):
        mean_time = np.mean(time_list) * 1000  # Convert to ms
        total_time = np.sum(time_list)
        print(f"\n{name}:")
        print(f"  Mean: {mean_time:.4f} ms | Total: {total_time:.6f} s")
    
    # Determine best filter for non-Gaussian noise
    rmse_dict = {'EKF-RHONN': metrics_ekf['RMSE_total'], 
                 'UKF-RHONN': metrics_ukf['RMSE_total'], 
                 'PF-RHONN': metrics_pf['RMSE_total']}
    best_filter = min(rmse_dict, key=rmse_dict.get)
    
    print("\n" + "="*70)
    print(f"BEST FILTER WITH NON-GAUSSIAN NOISE: {best_filter}")
    print(f"RMSE total: {rmse_dict[best_filter]:.6f}")
    print("="*70)
    
    # Robustness analysis
    print("\n" + "="*70)
    print("ROBUSTNESS ANALYSIS (Non-Gaussian vs Gaussian)")
    print("="*70)
    print("\nExpected behavior with non-Gaussian noise:")
    print("  1. EKF/UKF should degrade more (assume Gaussian noise)")
    print("  2. PF should be more robust (uses Student-t likelihood)")
    print("  3. Error metrics will be higher overall than with Gaussian noise")
    print("\nKurtosis > 3 indicates heavier tails than Gaussian")
    print("Impulsive noise causes occasional large outliers")

# ============================================================
# 8) VISUALIZATION WITH NON-GAUSSIAN NOISE
# ============================================================

print("\nGenerating visualizations with non-Gaussian noise...")

# Configuration for thesis-style plots
thesis_config = {
    'font_family': 'Computer Modern, serif',
    'font_size': 16,
    'title_font_size': 18,
    'legend_font_size': 14,
    'line_width_est': 2.0,
    'plot_width': 1000,
    'plot_height': 500,
    'grid_color': 'rgba(200, 200, 200, 0.3)',
    'grid_width': 0.5
}

# --- Non-Gaussian Noise Distribution Plots ---
fig_noise = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Process Noise: θ₁ (Laplace)', 
                    'Process Noise: ω₁ (Gaussian Mixture)',
                    'Measurement Noise: θ₁ (Student-t)', 
                    'Measurement Noise: ω₁ (Impulsive)'),
    vertical_spacing=0.15,
    horizontal_spacing=0.1
)

# Generate sample noise for visualization
n_samples = 1000
sample_proc_laplace = NonGaussianNoise.laplace_noise(scale=0.002, size=n_samples)
sample_proc_mixture = NonGaussianNoise.gaussian_mixture_noise(
    means=[0, 0], stds=[0.001, 0.01], weights=[0.9, 0.1], size=n_samples
)
sample_meas_student = NonGaussianNoise.student_t_noise(df=3, scale=0.015, size=n_samples)
sample_meas_impulsive = NonGaussianNoise.impulsive_noise(
    std_normal=0.02, std_impulse=0.2, impulse_prob=0.05, size=n_samples
)

# Add histograms
fig_noise.add_trace(go.Histogram(x=sample_proc_laplace, nbinsx=100, 
                                 name='Laplace', marker_color='#1f77b4'),
                   row=1, col=1)
fig_noise.add_trace(go.Histogram(x=sample_proc_mixture, nbinsx=100,
                                 name='Mixture', marker_color='#2ca02c'),
                   row=1, col=2)
fig_noise.add_trace(go.Histogram(x=sample_meas_student, nbinsx=100,
                                 name='Student-t', marker_color='#d62728'),
                   row=2, col=1)
fig_noise.add_trace(go.Histogram(x=sample_meas_impulsive, nbinsx=100,
                                 name='Impulsive', marker_color='#9467bd'),
                   row=2, col=2)

fig_noise.update_layout(
    title={'text': 'Non-Gaussian Noise Distributions', 'x': 0.5, 'xanchor': 'center'},
    showlegend=False,
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    paper_bgcolor='white',
    width=thesis_config['plot_width'],
    height=700,
    bargap=0.02
)

fig_noise.show()

# --- Training Times Plot ---
fig_times = go.Figure()
sim_time = t[1:]  # Exclude first time step

fig_times.add_trace(go.Scatter(
    x=sim_time, y=pf_training_times,
    mode='lines',
    name='PF-RHONN',
    line=dict(color='#d62728', width=2),
))

fig_times.update_layout(
    title={'text': 'Training Time per Iteration (Non-Gaussian Noise)', 'x': 0.5},
    xaxis_title='Simulation Time (s)',
    yaxis_title='Training Time (s)',
    xaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color'], 
               gridwidth=thesis_config['grid_width'], showline=True),
    yaxis=dict(type='log', showgrid=True, gridcolor=thesis_config['grid_color'],
               gridwidth=thesis_config['grid_width'], showline=True),
    legend=dict(x=0.02, y=0.98, bgcolor='rgba(255, 255, 255, 0.9)'),
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    width=thesis_config['plot_width'],
    height=thesis_config['plot_height']
)

fig_times.show()

# --- State Plots with Non-Gaussian Noise ---
states_info = [
    {'idx': 0, 'var': 'θ₁', 'desc': 'First Pendulum Angle', 'y_label': 'Angle θ₁ (rad)'},
    {'idx': 1, 'var': 'θ₂', 'desc': 'Second Pendulum Angle', 'y_label': 'Angle θ₂ (rad)'},
    {'idx': 2, 'var': 'ω₁', 'desc': 'Angular Velocity 1', 'y_label': 'Velocity ω₁ (rad/s)'},
    {'idx': 3, 'var': 'ω₂', 'desc': 'Angular Velocity 2', 'y_label': 'Velocity ω₂ (rad/s)'}
]

for state_info in states_info:
    i = state_info['idx']
    
    fig = go.Figure()
    
    # Measured state (with non-Gaussian noise)
    fig.add_trace(go.Scatter(
        x=t, y=y_measured[:, i],
        mode='lines',
        name='Measurement (non-Gaussian noise)',
        line=dict(color='#808080', width=1.5, dash='dot'),
        showlegend=True
    ))
    
    # True state (with process noise)
    fig.add_trace(go.Scatter(
        x=t, y=x_true[:, i],
        mode='lines',
        name='True State',
        line=dict(color='#000000', width=2.5),
        showlegend=True
    ))
    
    # Filter estimates
    fig.add_trace(go.Scatter(
        x=t, y=x_est_ekf[:, i], mode='lines',
        name='EKF-RHONN', line=dict(color='#1f77b4', width=thesis_config['line_width_est'], dash='dash')
    ))
    
    fig.add_trace(go.Scatter(
        x=t, y=x_est_ukf[:, i], mode='lines',
        name='UKF-RHONN', line=dict(color='#2ca02c', width=thesis_config['line_width_est'], dash='dot')
    ))
    
    fig.add_trace(go.Scatter(
        x=t, y=x_est_pf[:, i], mode='lines',
        name='PF-RHONN', line=dict(color='#d62728', width=thesis_config['line_width_est'], dash='dashdot')
    ))
    
    fig.update_layout(
        title={'text': f'State {state_info["var"]}: {state_info["desc"]} (Non-Gaussian Noise)', 'x': 0.5},
        xaxis_title='Time (s)',
        yaxis_title=state_info['y_label'],
        xaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color']),
        yaxis=dict(showgrid=True, gridcolor=thesis_config['grid_color']),
        legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='center', x=0.5),
        font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
        plot_bgcolor='white',
        width=thesis_config['plot_width'],
        height=thesis_config['plot_height']
    )
    
    fig.show()

# --- Metrics Comparison Bar Chart ---
fig_metrics = go.Figure()

metrics_names = ['RMSE', 'MAE', 'NRMSE']
ekf_vals = [metrics_ekf['RMSE_total'], metrics_ekf['MAE_total'], metrics_ekf['NRMSE_total']]
ukf_vals = [metrics_ukf['RMSE_total'], metrics_ukf['MAE_total'], metrics_ukf['NRMSE_total']]
pf_vals = [metrics_pf['RMSE_total'], metrics_pf['MAE_total'], metrics_pf['NRMSE_total']]

x = np.arange(len(metrics_names))
width = 0.25

fig_metrics.add_trace(go.Bar(
    x=x - width, y=ekf_vals, width=width,
    name='EKF-RHONN', marker_color='#1f77b4'
))

fig_metrics.add_trace(go.Bar(
    x=x, y=ukf_vals, width=width,
    name='UKF-RHONN', marker_color='#2ca02c'
))

fig_metrics.add_trace(go.Bar(
    x=x + width, y=pf_vals, width=width,
    name='PF-RHONN', marker_color='#d62728'
))

fig_metrics.update_layout(
    title={'text': 'Error Metrics Comparison (Non-Gaussian Noise)', 'x': 0.5},
    xaxis=dict(tickmode='array', tickvals=x, ticktext=metrics_names, showline=True),
    yaxis=dict(title='Metric Value', showgrid=True, gridcolor=thesis_config['grid_color']),
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='center', x=0.5),
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    width=thesis_config['plot_width'],
    height=thesis_config['plot_height'],
    barmode='group'
)

fig_metrics.show()

# --- Noise Time Series Visualization ---
fig_noise_ts = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Process Noise: θ₁', 'Process Noise: ω₁',
                    'Measurement Noise: θ₁', 'Measurement Noise: ω₁'),
    vertical_spacing=0.15,
    horizontal_spacing=0.1
)

# Process noise time series
fig_noise_ts.add_trace(go.Scatter(
    x=t[1:], y=process_noise_history[:, 0],
    mode='lines',
    line=dict(color='#1f77b4', width=1.5),
    name='θ₁ Process Noise'
), row=1, col=1)

fig_noise_ts.add_trace(go.Scatter(
    x=t[1:], y=process_noise_history[:, 2],
    mode='lines',
    line=dict(color='#2ca02c', width=1.5),
    name='ω₁ Process Noise'
), row=1, col=2)

# Measurement noise time series
fig_noise_ts.add_trace(go.Scatter(
    x=t, y=measurement_noise_history[:, 0],
    mode='lines',
    line=dict(color='#d62728', width=1.5),
    name='θ₁ Measurement Noise'
), row=2, col=1)

fig_noise_ts.add_trace(go.Scatter(
    x=t, y=measurement_noise_history[:, 2],
    mode='lines',
    line=dict(color='#9467bd', width=1.5),
    name='ω₁ Measurement Noise'
), row=2, col=2)

fig_noise_ts.update_layout(
    title={'text': 'Non-Gaussian Noise Time Series', 'x': 0.5, 'xanchor': 'center'},
    showlegend=False,
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    paper_bgcolor='white',
    width=thesis_config['plot_width'],
    height=700
)

fig_noise_ts.update_yaxes(title_text="Noise Amplitude", row=1, col=1)
fig_noise_ts.update_yaxes(title_text="Noise Amplitude", row=1, col=2)
fig_noise_ts.update_yaxes(title_text="Noise Amplitude", row=2, col=1)
fig_noise_ts.update_yaxes(title_text="Noise Amplitude", row=2, col=2)
fig_noise_ts.update_xaxes(title_text="Time (s)", row=2, col=1)
fig_noise_ts.update_xaxes(title_text="Time (s)", row=2, col=2)

fig_noise_ts.show()

print("\n✅ Simulation with NON-GAUSSIAN NOISE completed successfully.")
print("\n" + "="*70)
print("SUMMARY OF NON-GAUSSIAN NOISE EFFECTS")
print("="*70)
print("\n1. Process Noise:")
print("   - θ₁, θ₂: Laplace distribution (heavy tails)")
print("   - ω₁, ω₂: Gaussian mixture (10% impulses)")
print("\n2. Measurement Noise:")
print("   - θ₁, θ₂: Student's t-distribution (df=3, very heavy tails)")
print("   - ω₁, ω₂: Impulsive noise (5% large outliers)")
print("\n3. Expected Filter Behavior:")
print("   - EKF/UKF: Assumes Gaussian noise → Poor performance with outliers")
print("   - PF: Uses Student-t likelihood → More robust to heavy tails")
print("\n4. Key Statistics:")
print("   - Kurtosis > 3 indicates heavier tails than Gaussian")
print("   - Outliers (>3σ) are expected and intentionally included")
```
