```python
# Neural Identifier Training - 2-DOF Planar Manipulator
# Methods: EKF, UKF, Particle Filter

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python

# ============================================================
# 1) True nonlinear system (2-DOF Robot Arm)
# ============================================================
def plant_dynamics(state, u):
    """
    Dynamics of a 2-link planar robot arm.
    state = [q1, q2, dq1, dq2] (Angles and Velocities)
    u     = [tau1, tau2]       (Torques)
    """
    # Robot Parameters
    m1, m2 = 1.0, 1.0  # Mass (kg)
    l1, l2 = 1.0, 1.0  # Lengths (m)
    g = 9.81
    
    q1, q2, dq1, dq2 = state
    tau1, tau2 = u

    # --- Mass Matrix M(q) ---
    c2 = np.cos(q2)
    s2 = np.sin(q2)
    
    M11 = (m1 + m2) * l1**2 + m2 * l2**2 + 2 * m2 * l1 * l2 * c2
    M12 = m2 * l2**2 + m2 * l1 * l2 * c2
    M21 = M12
    M22 = m2 * l2**2
    M = np.array([[M11, M12], [M21, M22]])

    # --- Coriolis/Centrifugal Matrix C(q, dq) ---
    h = -m2 * l1 * l2 * s2
    C11 = h * dq2
    C12 = h * (dq1 + dq2)
    C21 = -h * dq1
    C22 = 0.0
    C = np.array([[C11, C12], [C21, C22]])

    # --- Gravity Vector G(q) ---
    s1 = np.sin(q1)
    s12 = np.sin(q1 + q2)
    G1 = (m1 + m2) * g * l1 * s1 + m2 * g * l2 * s12
    G2 = m2 * g * l2 * s12
    G = np.array([G1, G2])

    # --- Equation of Motion: M*ddq + C*dq + G = tau ---
    # ddq = inv(M) * (tau - C*dq - G)
    
    damping = 0.5 * np.array([dq1, dq2]) # Add some friction
    torque_vector = np.array([tau1, tau2])
    
    rhs = torque_vector - (C @ np.array([dq1, dq2])) - G - damping
    
    # Solve for accelerations
    ddq = np.linalg.solve(M, rhs)
    
    return np.concatenate(([dq1, dq2], ddq))
```

```python

def plant(x_k, u_k, dt=0.01, process_noise_std=1e-3):
    """
    Euler integration step.
    """
    x_dot = plant_dynamics(x_k, u_k)
    x_kp1 = x_k + dt * x_dot
    
    # Add small process noise
    noise = np.random.laplace(4) * process_noise_std
    return x_kp1 + noise
```

```python

# ============================================================
# 2) RHONN structure (Updated for Control Inputs)
# ============================================================
def sigmoidal(z, beta=1.0):
    """Sigmoid S(z). Beta reduced to widen the active range."""
    z = np.clip(z, -50, 50) 
    return 1.0 / (1.0 + np.exp(-beta * z))

def construct_z_vector(x_est, u_input):
    """
    Features for 2-DOF Arm.
    x_est = [q1, q2, dq1, dq2]
    u_input = [tau1, tau2]
    
    CRITICAL: The acceleration depends Linearly on Tau. 
    So Tau should be passed directly (or scaled), not inside a sigmoid.
    """
    q1, q2, dq1, dq2 = x_est
    tau1, tau2 = u_input
    
    # Scale inputs for sigmoids to avoid saturation
    s_q1 = sigmoidal(q1)
    s_q2 = sigmoidal(q2)
    s_dq1 = sigmoidal(dq1)
    s_dq2 = sigmoidal(dq2)
    
    # Interaction terms (physics based intuition)
    # Cosine terms are common in robotics, approximated by sigmoid interactions
    
    return np.array([
        # s_q1, s_q2, s_dq1, s_dq2,     # Individual states
        s_q2 * s_dq1,                 # Coriolis interaction proxy
        s_q2 * s_dq2,                 # Coriolis interaction proxy
        s_dq1**2, s_dq2**2,           # Centrifugal proxies
        tau1 * 0.1, tau2 * 0.1,       # Linear Inputs (Scaled)
        1.0                           # Bias
    ])
```

```python

# ============================================================
# 3) Trainers (EKF, UKF, PF) - Generic Classes
# ============================================================
# (Reusing the robust classes from previous steps, slightly adapted for input u)

class Generic_RHONN_Trainer:
    """ Base class to handle the loop logic easily """
    def get_prediction(self, weights, x_k, u_k):
        z = construct_z_vector(x_k, u_k)
        return np.dot(weights, z)

class EKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, n_weights, eta=1.0, P0=1.2, Q=1e-2, R=1e-3):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(n_weights)*0.1 for _ in range(n_neurons)]
        self.P = [np.eye(n_weights)*P0 for _ in range(n_neurons)]
        self.Q = np.eye(n_weights)*Q
        self.R = R
        self.eta = eta

    def update(self, x_kp1, x_k, u_k):
        z = construct_z_vector(x_k, u_k)
        H = z.reshape(-1, 1)
        
        for i in range(self.n_neurons):
            # Predict P
            P_pred = self.P[i] + self.Q
            
            # Kalman Gain
            S = self.R + (H.T @ P_pred @ H)[0,0]
            K = (P_pred @ H).flatten() / S
            
            # Error
            y_pred = np.dot(self.weights[i], z)
            err = x_kp1[i] - y_pred
            
            # Update Weights
            self.weights[i] += self.eta * K * err
            
            # Update Covariance (Joseph form)
            I_KH = np.eye(len(z)) - np.outer(K, z)
            self.P[i] = I_KH @ P_pred @ I_KH.T + np.outer(K, K)*self.R

class UKF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, n_weights, eta=2.7, alpha=8.02e-3):
        self.n_neurons = n_neurons
        self.weights = [np.random.randn(n_weights)*0.1 for _ in range(n_neurons)]
        self.P = [np.eye(n_weights) for _ in range(n_neurons)]
        self.Q = np.eye(n_weights)*1e-3
        self.R = 1e-6
        self.eta = eta
        
        # Sigma params
        self.n = n_weights
        self.lambda_ = alpha**2 * (self.n) - self.n
        self.Wm = np.full(2*self.n+1, 1/(2*(self.n+self.lambda_)))
        self.Wc = np.copy(self.Wm)
        self.Wm[0] = self.lambda_/(self.n+self.lambda_)
        self.Wc[0] = self.Wm[0] + (3 - alpha**2)

    def update(self, x_kp1, x_k, u_k):
        z = construct_z_vector(x_k, u_k)
        
        for i in range(self.n_neurons):
            # Generate Sigmas
            try:
                L = np.linalg.cholesky((self.n + self.lambda_) * self.P[i])
            except:
                L = np.eye(self.n) * 0.1 # Fallback
                
            sigmas = np.zeros((2*self.n+1, self.n))
            sigmas[0] = self.weights[i]
            for k in range(self.n):
                sigmas[k+1] = self.weights[i] + L[:,k]
                sigmas[self.n+k+1] = self.weights[i] - L[:,k]
            
            # Transform (Measurement is linear: w^T * z)
            Y_sigmas = np.dot(sigmas, z)
            y_mean = np.sum(self.Wm * Y_sigmas)
            
            # Covariances
            Py = np.sum(self.Wc * (Y_sigmas - y_mean)**2) + self.R
            Pxy = np.zeros(self.n)
            for k in range(2*self.n+1):
                Pxy += self.Wc[k] * (sigmas[k] - self.weights[i]) * (Y_sigmas[k] - y_mean)
                
            # Update
            K = Pxy / Py
            err = x_kp1[i] - y_mean
            self.weights[i] += self.eta * K * err
            self.P[i] -= np.outer(K, K) * Py
            
            # Regularize P
            self.P[i] += np.eye(self.n)*1e-6


class PF_Trainer(Generic_RHONN_Trainer):
    def __init__(self, n_neurons, n_weights, n_particles=300, Q_std=None, R_std=None, use_optimization=False):
        """
        Args:
            n_neurons: Number of states to estimate
            n_weights: Number of features in the RHONN
            n_particles: Number of particles per state
            Q_std: Process noise std per state (array or scalar)
            R_std: Measurement noise std per state (array or scalar)
            use_optimization: If True, apply L-BFGS-B optimization after PF update
        """
        self.n_neurons = n_neurons
        self.n_weights = n_weights
        self.n_particles = n_particles
        self.use_optimization = use_optimization
        self.particles = [np.random.randn(n_particles, n_weights)*0.2 for _ in range(n_neurons)]
        self.weights_pf = [np.ones(n_particles)/n_particles for _ in range(n_neurons)]
        
        # Set Q_std per state (default 0.6 for all states)
        if Q_std is None:
            self.Q_std = np.ones(n_neurons) * 0.6
        elif np.isscalar(Q_std):
            self.Q_std = np.ones(n_neurons) * Q_std
        else:
            self.Q_std = np.array(Q_std)
            assert len(self.Q_std) == n_neurons, "Q_std length must match n_neurons"
        
        # Set R_std per state (default 0.2 for all states)
        if R_std is None:
            self.R_std = np.ones(n_neurons) * 0.2
        elif np.isscalar(R_std):
            self.R_std = np.ones(n_neurons) * R_std
        else:
            self.R_std = np.array(R_std)
            assert len(self.R_std) == n_neurons, "R_std length must match n_neurons"

    def update(self, x_kp1, x_k, u_k):
        z = construct_z_vector(x_k, u_k)
        
        for i in range(self.n_neurons):
            # 1. Drift (use per-state Q_std)
            self.particles[i] += np.random.randn(self.n_particles, self.n_weights) * self.Q_std[i]
            
            # 2. Weight (use per-state R_std) - Log-likelihood approach
            preds = self.particles[i] @ z
            err = x_kp1[i] - preds
            log_likelihood = -0.5 * (err / self.R_std[i])**2
            
            # Update weights in log-space for numerical stability
            log_weights = np.log(self.weights_pf[i] + 1e-300) + log_likelihood
            
            # Normalize using log-sum-exp trick
            max_log_weight = np.max(log_weights)
            log_weights_shifted = log_weights - max_log_weight
            weights_unnormalized = np.exp(log_weights_shifted)
            self.weights_pf[i] = weights_unnormalized / np.sum(weights_unnormalized)
            
            # 3. Resample (Simple systematic)
            # Systematic resampling
            eff_N = 1.0 / np.sum(self.weights_pf[i]**2)
            if eff_N < self.n_particles/2:
                # Systematic resampling algorithm
                positions = (np.arange(self.n_particles) + np.random.uniform()) / self.n_particles
                cumulative_sum = np.cumsum(self.weights_pf[i])
                indices = np.searchsorted(cumulative_sum, positions)
                
                self.particles[i] = self.particles[i][indices]
                self.weights_pf[i].fill(1.0/self.n_particles)
            
            # 4. Optional: Apply optimization to refine estimate
            if self.use_optimization:
                # Get weighted average
                w_avg = np.average(self.particles[i], axis=0, weights=self.weights_pf[i])
                # Optimize it
                w_optimized = self.optimize_estimate(i, x_kp1[i], z, w_avg)
                # Replace worst particles with optimized solution
                worst_idx = np.argmin(self.weights_pf[i])
                self.particles[i][worst_idx] = w_optimized
    
    def optimize_estimate(self, neuron_idx, target_value, z_vector, initial_weights=None):
        """
        Optional: Refine particle filter estimate using L-BFGS-B optimization.
        Minimizes squared prediction error.
        
        Args:
            neuron_idx: Index of the neuron to optimize
            target_value: Target output value (measurement)
            z_vector: Input feature vector
            initial_weights: Starting point (if None, uses weighted average)
        
        Returns:
            Optimized weight vector
        """
        from scipy.optimize import minimize
        
        # Use weighted average as initial guess
        if initial_weights is None:
            initial_weights = np.average(self.particles[neuron_idx], 
                                        axis=0, 
                                        weights=self.weights_pf[neuron_idx])
        
        # Objective function: squared prediction error
        def objective(w):
            prediction = np.dot(w, z_vector)
            error = target_value - prediction
            return error**2
        
        # Gradient of objective
        def gradient(w):
            prediction = np.dot(w, z_vector)
            error = target_value - prediction
            return -2 * error * z_vector
        
        # Optimize using L-BFGS-B
        result = minimize(
            objective,
            initial_weights,
            method='L-BFGS-B',
            jac=gradient,
            options={'maxiter': 50, 'ftol': 1e-6}
        )
        
        return result.x if result.success else initial_weights
                
    def get_estimates(self):
        return [np.average(self.particles[i], axis=0, weights=self.weights_pf[i]) for i in range(self.n_neurons)]
```

```python
# ============================================================
# 4) Simulation Main Loop - 2-DOF Manipulator
# ============================================================
if __name__ == "__main__":
    # Set random seed for reproducibility and fair comparison
    np.random.seed(42)
    
    n_steps = 1000
    dt = 0.01
    t = np.linspace(0, (n_steps-1)*dt, n_steps)
    
    # 4 States [q1, q2, dq1, dq2] - 7 Features from construct_z_vector
    n_states = 4
    n_features = 7
    
    # Pre-generate ALL noise sequences for fair comparison
    print("Pre-generating noise sequences for fair comparison...")
    process_noise_seq = np.random.laplace(4, size=(n_steps-1, 4)) * 1e-3
    measurement_noise_seq = np.random.randn(n_steps-1, 4) * 0.01
    
    # Generate true trajectory with pre-generated noise
    x_true = np.zeros((n_steps, 4))
    x_true[0] = [-np.pi/2, 0, 0, 0]
    
    # Excitation Input (Sine waves to excite dynamics)
    u_hist = np.zeros((n_steps, 2))
    for k in range(n_steps):
        # Time-varying torque to excite dynamics
        tau1 = 30.0 * np.sin(2.0 * t[k]) 
        tau2 = 15.0 * np.cos(3.0 * t[k])
        u_hist[k] = [tau1, tau2]
    
    # Generate true trajectory
    print("Generating true trajectory...")
    for k in range(n_steps - 1):
        x_dot = plant_dynamics(x_true[k], u_hist[k])
        x_true[k+1] = x_true[k] + dt * x_dot + process_noise_seq[k]
    
    # Initialize Trainers with same seed for fair comparison
    print("\nInitializing filters...")
    np.random.seed(7517)
    ekf = EKF_Trainer(n_states, n_features, eta=1.0)
    ukf = UKF_Trainer(n_states, n_features, eta=1.0, alpha=1e-2)
    
    pf_Q_std = [0.3, 0.3, 0.6, 0.6]  # Lower for angles, higher for velocities
    pf_R_std = [0.01, 0.01, 0.01, 0.01]  # Measurement noise
    pf = PF_Trainer(n_states, n_features, n_particles=1000, Q_std=pf_Q_std, R_std=pf_R_std, use_optimization=False)
    
    # Optional: PF with optimization (uncomment to enable)
    # pf_opt = PF_Trainer(n_states, n_features, n_particles=1000, Q_std=pf_Q_std, R_std=pf_R_std, use_optimization=True)
    use_pf_optimization = False  # Set to True to compare PF vs PF-OPT
    
    if use_pf_optimization:
        np.random.seed(7517)
        pf_opt = PF_Trainer(n_states, n_features, n_particles=1000, Q_std=pf_Q_std, R_std=pf_R_std, use_optimization=True)
        print("✓ PF with L-BFGS-B optimization enabled")
    
    # Arrays for estimates
    x_est_ekf = np.zeros((n_steps, 4))
    x_est_ukf = np.zeros((n_steps, 4))
    x_est_pf = np.zeros((n_steps, 4))
    
    x_est_ekf[0] = x_true[0]
    x_est_ukf[0] = x_true[0]
    x_est_pf[0] = x_true[0]
    
    if use_pf_optimization:
        x_est_pf_opt = np.zeros((n_steps, 4))
        x_est_pf_opt[0] = x_true[0]

    print("Training filters with identical noise sequences...")
    
    # Main simulation loop - Series-Parallel Architecture
    for k in range(n_steps - 1):
        # Use pre-generated measurement noise (same for all filters)
        x_measured = x_true[k+1] + measurement_noise_seq[k]
        
        # Construct feature vector (same for all filters)
        z = construct_z_vector(x_true[k], u_hist[k])
        
        # EKF Update & Predict
        ekf.update(x_measured, x_true[k], u_hist[k])
        for i in range(4):
            x_est_ekf[k+1, i] = np.dot(ekf.weights[i], z)
            
        # UKF Update & Predict
        ukf.update(x_measured, x_true[k], u_hist[k])
        for i in range(4):
            x_est_ukf[k+1, i] = np.dot(ukf.weights[i], z)
            
        # PF Update & Predict
        pf.update(x_measured, x_true[k], u_hist[k])
        w_pf = pf.get_estimates()
        for i in range(4):
            x_est_pf[k+1, i] = np.dot(w_pf[i], z)
        
        # PF with Optimization (if enabled)
        if use_pf_optimization:
            pf_opt.update(x_measured, x_true[k], u_hist[k])
            w_pf_opt = pf_opt.get_estimates()
            for i in range(4):
                x_est_pf_opt[k+1, i] = np.dot(w_pf_opt[i], z)
            
        if k % 100 == 0:
            status = f"Step {k}/{n_steps-1}"
            if use_pf_optimization:
                status += " (PF-OPT active)"
            print(status)

    # ============================================================
    # 5) Calculate Performance Metrics
    # ============================================================
    
    print("\nCalculating performance metrics...")
    
    # MSE per state
    mse_q1_ekf = np.mean((x_true[:, 0] - x_est_ekf[:, 0])**2)
    mse_q2_ekf = np.mean((x_true[:, 1] - x_est_ekf[:, 1])**2)
    mse_dq1_ekf = np.mean((x_true[:, 2] - x_est_ekf[:, 2])**2)
    mse_dq2_ekf = np.mean((x_true[:, 3] - x_est_ekf[:, 3])**2)
    
    mse_q1_ukf = np.mean((x_true[:, 0] - x_est_ukf[:, 0])**2)
    mse_q2_ukf = np.mean((x_true[:, 1] - x_est_ukf[:, 1])**2)
    mse_dq1_ukf = np.mean((x_true[:, 2] - x_est_ukf[:, 2])**2)
    mse_dq2_ukf = np.mean((x_true[:, 3] - x_est_ukf[:, 3])**2)
    
    mse_q1_pf = np.mean((x_true[:, 0] - x_est_pf[:, 0])**2)
    mse_q2_pf = np.mean((x_true[:, 1] - x_est_pf[:, 1])**2)
    mse_dq1_pf = np.mean((x_true[:, 2] - x_est_pf[:, 2])**2)
    mse_dq2_pf = np.mean((x_true[:, 3] - x_est_pf[:, 3])**2)
    
    mse_total_ekf = mse_q1_ekf + mse_q2_ekf + mse_dq1_ekf + mse_dq2_ekf
    mse_total_ukf = mse_q1_ukf + mse_q2_ukf + mse_dq1_ukf + mse_dq2_ukf
    mse_total_pf = mse_q1_pf + mse_q2_pf + mse_dq1_pf + mse_dq2_pf
    
    # Calculate MSE for PF-OPT if enabled
    if use_pf_optimization:
        mse_q1_pf_opt = np.mean((x_true[:, 0] - x_est_pf_opt[:, 0])**2)
        mse_q2_pf_opt = np.mean((x_true[:, 1] - x_est_pf_opt[:, 1])**2)
        mse_dq1_pf_opt = np.mean((x_true[:, 2] - x_est_pf_opt[:, 2])**2)
        mse_dq2_pf_opt = np.mean((x_true[:, 3] - x_est_pf_opt[:, 3])**2)
        mse_total_pf_opt = mse_q1_pf_opt + mse_q2_pf_opt + mse_dq1_pf_opt + mse_dq2_pf_opt
    
    # Print results
    print("\n" + "="*70)
    print("🏆 MEJOR FILTRO: ", end="")
    mse_dict = {'EKF': mse_total_ekf, 'UKF': mse_total_ukf, 'PF': mse_total_pf}
    if use_pf_optimization:
        mse_dict['PF-OPT'] = mse_total_pf_opt
    best_filter = min(mse_dict, key=mse_dict.get)
    print(f"{best_filter} (MSE total: {mse_dict[best_filter]:.6f})")
    print("="*70)
    
    print("\n--- Comparación de Desempeño (MSE) - Manipulador 2-DOF ---")
    print(f"EKF MSE q₁:   {mse_q1_ekf:.6f} | MSE dq₁:   {mse_dq1_ekf:.6f}")
    print(f"EKF MSE q₂:   {mse_q2_ekf:.6f} | MSE dq₂:   {mse_dq2_ekf:.6f}")
    print(f"UKF MSE q₁:   {mse_q1_ukf:.6f} | MSE dq₁:   {mse_dq1_ukf:.6f}")
    print(f"UKF MSE q₂:   {mse_q2_ukf:.6f} | MSE dq₂:   {mse_dq2_ukf:.6f}")
    print(f"PF  MSE q₁:   {mse_q1_pf:.6f} | MSE dq₁:   {mse_dq1_pf:.6f}")
    print(f"PF  MSE q₂:   {mse_q2_pf:.6f} | MSE dq₂:   {mse_dq2_pf:.6f}")
    
    if use_pf_optimization:
        print(f"\n--- PF con Optimización L-BFGS-B ---")
        print(f"PF-OPT MSE q₁: {mse_q1_pf_opt:.6f} | MSE dq₁: {mse_dq1_pf_opt:.6f}")
        print(f"PF-OPT MSE q₂: {mse_q2_pf_opt:.6f} | MSE dq₂: {mse_dq2_pf_opt:.6f}")
        print(f"Mejora vs PF: {((mse_total_pf - mse_total_pf_opt)/mse_total_pf)*100:.2f}%")
    
    # Configuration for thesis-style plots
    thesis_config = {
        'font_family': 'Computer Modern, serif',
        'font_size': 14,
        'title_font_size': 16,
        'legend_font_size': 12,
        'line_width_true': 2.5,
        'line_width_est': 2.0,
        'plot_width': 1000,
        'plot_height': 500,
        'grid_color': 'rgba(200, 200, 200, 0.3)',
        'grid_width': 0.5
    }
```

```python
# ============================================================
# Gráfica de Métricas de Error (RMSE, MAE, NRMSE)
# ============================================================

print("\nCalculando métricas de error adicionales...")

# Calcular errores para cada filtro
errors_ekf = x_true - x_est_ekf
errors_ukf = x_true - x_est_ukf
errors_pf = x_true - x_est_pf

# RMSE (Root Mean Square Error)
rmse_ekf = np.sqrt(np.mean(errors_ekf**2))
rmse_ukf = np.sqrt(np.mean(errors_ukf**2))
rmse_pf = np.sqrt(np.mean(errors_pf**2))

# MAE (Mean Absolute Error)
mae_ekf = np.mean(np.abs(errors_ekf))
mae_ukf = np.mean(np.abs(errors_ukf))
mae_pf = np.mean(np.abs(errors_pf))

# NRMSE (Normalized RMSE) - normalizado por el rango de los datos
data_range = np.max(x_true) - np.min(x_true)
nrmse_ekf = rmse_ekf / data_range
nrmse_ukf = rmse_ukf / data_range
nrmse_pf = rmse_pf / data_range

print(f"\nMétricas de Error Globales:")
print(f"EKF-RHONN: RMSE={rmse_ekf:.4f}, MAE={mae_ekf:.4f}, NRMSE={nrmse_ekf:.4f}")
print(f"UKF-RHONN: RMSE={rmse_ukf:.4f}, MAE={mae_ukf:.4f}, NRMSE={nrmse_ukf:.4f}")
print(f"PF-RHONN:  RMSE={rmse_pf:.4f}, MAE={mae_pf:.4f}, NRMSE={nrmse_pf:.4f}")

# Crear gráfica de barras agrupadas
fig_metrics = go.Figure()

filters = ['EKF-RHONN', 'UKF-RHONN', 'PF-RHONN']

fig_metrics.add_trace(go.Bar(
    name='RMSE',
    x=filters,
    y=[rmse_ekf, rmse_ukf, rmse_pf],
    marker_color='#5470C6',
    text=[f'{rmse_ekf:.4f}', f'{rmse_ukf:.4f}', f'{rmse_pf:.4f}'],
    textposition='outside',
    textfont=dict(size=12)
))

fig_metrics.add_trace(go.Bar(
    name='MAE',
    x=filters,
    y=[mae_ekf, mae_ukf, mae_pf],
    marker_color='#91CC75',
    text=[f'{mae_ekf:.4f}', f'{mae_ukf:.4f}', f'{mae_pf:.4f}'],
    textposition='outside',
    textfont=dict(size=12)
))

fig_metrics.add_trace(go.Bar(
    name='NRMSE',
    x=filters,
    y=[nrmse_ekf, nrmse_ukf, nrmse_pf],
    marker_color='#5470C6',
    text=[f'{nrmse_ekf:.4f}', f'{nrmse_ukf:.4f}', f'{nrmse_pf:.4f}'],
    textposition='outside',
    textfont=dict(size=12),
    opacity=0.7
))

fig_metrics.update_layout(
    title={
        'text': 'Métricas de Error',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': thesis_config['title_font_size'], 'family': thesis_config['font_family']}
    },
    xaxis_title='',
    yaxis_title='Valor de la Métrica',
    yaxis=dict(
        showgrid=True,
        gridcolor=thesis_config['grid_color'],
        gridwidth=thesis_config['grid_width'],
        showline=True,
        linewidth=1.5,
        linecolor='black',
        mirror=True,
        ticks='outside',
        tickwidth=1.5,
        ticklen=5
    ),
    xaxis=dict(
        showline=True,
        linewidth=1.5,
        linecolor='black',
        mirror=True,
        ticks='outside',
        tickwidth=1.5,
        ticklen=5
    ),
    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=1.02,
        xanchor='center',
        x=0.5,
        bgcolor='rgba(255, 255, 255, 0)',
        font=dict(size=thesis_config['legend_font_size'], family=thesis_config['font_family'])
    ),
    barmode='group',
    font=dict(size=thesis_config['font_size'], family=thesis_config['font_family']),
    plot_bgcolor='white',
    paper_bgcolor='white',
    width=thesis_config['plot_width'],
    height=thesis_config['plot_height'],
    margin=dict(l=80, r=40, t=100, b=60)
)

fig_metrics.show()
```
